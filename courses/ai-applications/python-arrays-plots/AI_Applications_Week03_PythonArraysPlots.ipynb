## 2. 연산과 검증

같은 위치의 원소를 곱하는 A*B와 행·열을 곱해 더하는 A@B를 계산해 비교합니다. 이어 A의 Inverse를 구하고, A@Inverse가 Identity가 되는지 확인합니다.

### A. 기준 예제 · 먼저 읽고 실행

```python
A=np.array([[2.,1.],[1.,3.]])
B=np.array([[1.,0.],[2.,1.]])
print('elementwise:',A*B,'matrix:',A@B)
print('inverse:',np.linalg.inv(A))
assert np.allclose(A@np.linalg.inv(A),np.eye(2))
values=np.array([3.,7.,5.]);print('max:',values.max(),'index:',values.argmax())
```

### A. 기준 예제 · 먼저 읽고 실행

```python
x=np.linspace(0,2*np.pi,100)
y=np.sin(x)
fig,ax=plt.subplots();ax.plot(x,y,label='sin(x)');i=y.argmax()
ax.annotate('sample maximum',(x[i],y[i]),xytext=(2,0.5),arrowprops={'arrowstyle':'->'})
ax.set(xlabel='x (rad)',ylabel='value',title='Sampled sine');ax.grid();ax.legend();plt.show()
```
