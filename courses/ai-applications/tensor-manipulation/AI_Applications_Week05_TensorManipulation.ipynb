### A. 기준 예제 · 먼저 읽고 실행

```python
A=torch.arange(6,dtype=torch.float32).reshape(2,3)
B=torch.arange(12,dtype=torch.float32).reshape(3,4)
print((A@B).shape)
print((A+torch.tensor([10,20,30])).shape)
assert (A@B).shape==(2,4)
```
