## 1. 간단한 CNN으로 각 층의 shape 확인하기

먼저 Conv Layer가 하나인 모델을 실행한 뒤, Conv Layer가 세 개인 CNN을 직접 만들어봅니다.

### A. 기준 예제 · 먼저 읽고 실행

```python
import sys, numpy as np, matplotlib.pyplot as plt
import torch
from torch import nn
import torch.nn.functional as F
torch.manual_seed(42)
torch.set_num_threads(2)
print('Python:', sys.version.split()[0], '| PyTorch:', torch.__version__)
small=nn.Sequential(nn.Conv2d(1,4,3,padding=1),nn.ReLU(),nn.MaxPool2d(2),nn.Flatten(1),nn.Linear(4*14*14,10))
x=torch.randn(8,1,28,28)
for layer in small:
    x=layer(x);print(type(layer).__name__,tuple(x.shape))
assert x.shape==(8,10)
```

### B. 직접 수행

## 2. Batch(배치) 크기와 parameter

### A. 기준 예제 · 먼저 읽고 실행

```python
for batch in [1,8,32]:
    print('batch:',batch,'output:',small(torch.randn(batch,1,28,28)).shape)
print('parameters:',sum(p.numel() for p in small.parameters()))
```
