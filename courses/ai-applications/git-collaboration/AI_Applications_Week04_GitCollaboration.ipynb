## 1. 내 작업 공간에서 Git 연습하기

임시 폴더에서 commit과 branch만 연습합니다. 실제 협업 과제는 과제 안내에서 이어갑니다.

### A. 기준 예제 · 먼저 읽고 실행

```python
import tempfile, pathlib, subprocess
repo=pathlib.Path(tempfile.mkdtemp(prefix='course-git-'))
def git(*args):
    return subprocess.check_output(['git','-C',str(repo),*args],text=True,stderr=subprocess.STDOUT)
git('init','-b','main');git('config','user.name','Student Example');git('config','user.email','example@example.invalid')
(repo/'README.md').write_text('# Practice\nBaseline\n')
git('add','README.md');git('commit','-m','Add baseline')
git('switch','-c','practice-change')
(repo/'README.md').write_text('# Practice\nBaseline\nMy change\n')
print(git('diff'));git('add','README.md');git('commit','-m','Describe one change')
print(git('log','--oneline','--all','--graph'))
```

### B. 직접 수행
