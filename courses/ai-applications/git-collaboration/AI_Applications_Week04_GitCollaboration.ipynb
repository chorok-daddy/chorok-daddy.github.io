혼자라면 main에 직접 수정·Commit할 수 있습니다. 작은 팀에서는 친구를 Collaborator로 초대해 수정 권한을 공유할 수도 있습니다. 항상 별도의 검토 요청을 거쳐야 하는 것은 아닙니다.

반면 처음 보는 사용자의 개선은 받아보고 싶어도, 기본 버전을 직접 수정할 권한까지 줄 필요는 없습니다. 공개 실습 코드의 README에서 빠진 준비 단계를 발견한 외부 사용자가 되었다고 생각해보세요.

1. **원본 main 확인:** 현재 안내에서 빠진 내용을 확인합니다. Repository는 파일과 변경 이력을 보관하는 저장소입니다.
2. **Fork·Branch:** 원본을 내 계정의 별도 Repository로 복사(Fork)하고 practice Branch에서 수정합니다. 원본은 아직 바뀌지 않습니다.
3. **Commit:** “실행 전 준비 단계 추가”라는 설명과 함께 변경을 기록합니다.
4. **Pull Request(PR)·검토:** 내 사본의 practice에서 원본 main으로 변경을 제안합니다. 관리자가 보완을 요청하면 같은 Branch에서 수정·Commit합니다. GitHub에 반영된 추가 변경도 같은 PR에서 검토합니다.
5. **Merge·확인:** 관리자가 원본 main에 합치면 모두가 개선된 안내를 사용합니다. 다음 작업 때는 내 사본을 원본의 최신 변경에 맞춘 뒤 새 Branch에서 시작합니다.

**필수 실습에서는** 본인 Repository에서 Branch·Commit·PR·Merge를 수행하며 수정자와 검토자의 역할을 모두 경험합니다. Fork는 필요하지 않습니다. 동료 Repository를 Fork해 원본에 PR을 제안하는 과정은 선택 실습입니다. 아는 팀이라도 여러 사람이 쓰는 기본 버전에는 이 검토 흐름이 유용합니다.

**Colab과 GitHub의 역할:** 아래 Colab 코드는 임시 Repository에서 수정·Commit·Branch를 연습합니다. 이어서 GitHub 웹에서 별도 Repository를 만들어 PR·Merge까지 완성합니다. Colab 예제가 GitHub에 자동 업로드되지는 않습니다. 로컬 Commit을 GitHub에 보내는 작업은 Push이며, 이번처럼 GitHub 웹에서 직접 만든 Commit에는 별도 로컬 Push가 필요하지 않습니다.

## 1. 내 작업 공간에서 Git 연습하기

임시 폴더에서 commit과 branch만 연습합니다. 실제 협업 과제는 과제 안내에서 이어갑니다.

### A. 기준 예제 · 먼저 읽고 실행

```python
import tempfile, pathlib, subprocess
repo=pathlib.Path(tempfile.mkdtemp(prefix='course-git-'))
def git(*args):
    return subprocess.check_output(['git','-C',str(repo),*args],text=True,stderr=subprocess.STDOUT)
git('init','-b','main');git('config','user.name','Student Example');git('config','user.email','example@example.invalid')
(repo/'README.md').write_text('# Practice\nBaseline\n')
git('add','README.md');git('commit','-m','Add baseline')
git('switch','-c','practice-change')
(repo/'README.md').write_text('# Practice\nBaseline\nMy change\n')
print(git('diff'));git('add','README.md');git('commit','-m','Describe one change')
print(git('log','--oneline','--all','--graph'))
```

### B. 직접 수행
