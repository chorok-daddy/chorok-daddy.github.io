After obtaining probabilities, use a Threshold to decide whether to prioritize reducing missed detections or false alarms.

## Score, probability, and decision {#점수확률판정}

Applying sigmoid to logit z gives p=1/(1+exp(−z)). Probability and the final decision are different. Lowering the threshold yields more positive predictions but can also increase false alarms.

$$p=\frac{1}{1+e^{-z}}$$

## Values passed to the loss function {#loss-function에-넣는-값}

Pass logits before Sigmoid to BCEWithLogitsLoss. Evaluate the training loss separately from the costs of misses and false alarms in deployment.

## Worked example {#풀이-예제}

$\mathrm{logit}=\begin{bmatrix}-2&-0.5&0&0.5&2\end{bmatrix}$, $\mathrm{Label}=\begin{bmatrix}0&0&1&1&1\end{bmatrix}$.

```python
import numpy as np
import matplotlib.pyplot as plt
logit=np.array([-2,-.5,0,.5,2]);y=np.array([0,0,1,1,1]);p=1/(1+np.exp(-logit))
print(p,p>=.5)
```
