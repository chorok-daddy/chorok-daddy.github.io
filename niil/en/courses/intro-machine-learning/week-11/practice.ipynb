## Cluster Count and Inertia {#군집-수와-관성값}

### Starting conditions {#시작-조건}

Use make_blobs(n_samples=180,centers=3,cluster_std=.8,random_state=7) and $K\in\{2,3,4\}$.

### Procedure {#진행-방법}

1.  Train KMeans(random_state=7,n_init=10) for each K.
2.  Plot cluster centers and assignments, and record Inertia and Silhouette.

### Results to retain {#남길-결과}

- Three cluster plots and a metrics table.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
X,_=make_blobs(n_samples=180,centers=3,cluster_std=.8,random_state=7)
model=KMeans(n_clusters=3,random_state=7,n_init=10).fit(X)
print(model.inertia_,silhouette_score(X,model.labels_))
```
