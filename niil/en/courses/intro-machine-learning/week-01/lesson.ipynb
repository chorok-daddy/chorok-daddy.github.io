## Worked example

$x=\begin{bmatrix}1&2&3&4&5&6&7&8\end{bmatrix}$, $y=\begin{bmatrix}0&0&0&0&1&1&1&1\end{bmatrix}$. x is a sensor feature, and y contains Labels for two states.

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(1,9);y=(x>4).astype(int)
pred=(x>4).astype(int)
print(np.mean(pred==y))
```
