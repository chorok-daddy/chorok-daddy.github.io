## 🚦 Module 1: Code That Follows Conditions --- if Statements {#vertical_traffic_light-모듈-1-조건을-따라-움직이는-코드-조건문if}

- **What is a conditional statement?**

  - Syntax that executes a block of code only when a specified condition is True.

- **Basic structure**

  ``` {.sourceCode .python}
  if 조건식:
      # Code to run when the condition is True
  elif 다른_조건식:
      # Code to run when the other condition is True
  else:
      # Code to run when all conditions above are False
  ```

- Applications in machine learning

  - ex1) \"If the rating is at least 9.0, classify it as a recommended movie.\"
  - ex2) \"If the data contain missing values, fill them with a specified value.\"
  - Essential for data preprocessing and defining model rules.

````python
# Example: Decide whether to recommend a movie based on its rating
movie_rating = 9.1

# 'If' movie_rating is greater than or equal to 9.0
if movie_rating >= 9.0:
    print("✨ 강력 추천하는 영화입니다!")
# 'Otherwise'
else:
    print("볼만한 영화입니다.")
````

````python
# Example: Decide whether to recommend a movie based on its rating
movie_rating = 8.5

# 'If' movie_rating is greater than or equal to 9.0
if movie_rating >= 9.0:
    print("✨ 강력 추천하는 영화입니다!")
# 'Otherwise'
else:
    print("볼만한 영화입니다.")

# Change movie_rating to 8.5 and run again!
````

````python
# A more detailed recommendation system
movie_rating = 8.8

if movie_rating >= 9.0:
    print("✨ 강력 추천! (9.0점 이상)")
# 'Else if' the rating is 8.5 or higher
elif movie_rating >= 8.5:
    print("👍 추천! (8.5점 이상)")
# 'In all other cases'
else:
    print("🤔 보통. (8.5점 미만)")
````

## 🔁 Module 2: All the Data at Once! for Loops {#repeat-모듈-2-모든-데이터를-한-번에-반복문for}

- **What Is a Loop?**

  - For all data in a structure such as a list or dictionary,
  - it **repeats the same task in sequence**.

- **Basic structure**

  ``` {.sourceCode .python}
  for 아이템을_담을_변수 in 데이터_리스트:
      # Code to run for each list item
  ```

- Uses in Machine Learning

  - Essential for large-scale data processing, such as loading and processing millions of images individually,
  - or calculating the mean or sum of an entire dataset.

````python
# The movie dataset created in Week 3
my_movie_dataset = [
    { "title": "인셉션",
     "director": "크리스토퍼 놀란",
      "rating": 9.29 },
    { "title": "기생충",
     "director": "봉준호",
      "rating": 9.07 },
    { "title": "어바웃 타임",
     "director": "리차드 커티스",
      "rating": 9.27 }
]

# Iterate through the dataset (list), placing each movie's information (dictionary) in movie
print("--- 모든 영화 제목 출력 ---")
for movie in my_movie_dataset:
    # Retrieve and print the value of the 'title' key from each movie (dictionary)
    print(movie['title'])
````

````python
# Combine loops and conditionals to select only 'movies rated 9.2 or higher'

print("\n--- 평점 9.2 이상인 추천 영화 ---")
for movie in my_movie_dataset:
    # 'If' the current movie's rating is greater than or equal to 9.2
    if movie['rating'] >= 9.2:
        # Print that movie's title and rating
        print(f"제목: {movie['title']}, 평점: {movie['rating']}")
````

## ⚙️ Module 3: Reusable Code Components --- Functions (def) {#gear-모듈-3-나만의-코드-부품-함수def}

- **What is a function?**

  - A block of code that performs a specific task
  - and can be **named and reused**.

- **Basic structure**

  ``` {.sourceCode .python}
  def 함수이름(입력값1, 입력값2, ...):
      # Code to run
      return 결과값
  ```

- Applications in machine learning

  - Implement data preprocessing as a function to apply the same process to multiple datasets.

  - A model itself can be viewed as a function.

    \[Input (data) → Processing (prediction) → Output (result)\]

  - Organize complex code by functionality to make it easier to maintain.

````python
# A simple function that adds two numbers
def add_numbers(x, y):
    """
    이 함수는 두 개의 숫자 x와 y를 입력받아,
    그 합을 계산하여 결과값으로 돌려줍니다.
    """
    result = x + y
    return result

# Call (use) the function
sum_result = add_numbers(5, 3)
print(f"5와 3의 합은? {sum_result}")

another_sum = add_numbers(100, 200)
print(f"100과 200의 합은? {another_sum}")
````

````python
# Turn Module 2's 'rating-based movie recommendation' logic into a function.

def recommend_movies(dataset, min_rating):
    """
    영화 데이터셋과 최소 평점을 입력받아,
    조건에 맞는 영화 제목 리스트를 결과로 돌려줍니다.
    """
    recommended_list = [] # An empty list for recommended movies
    for movie in dataset:
        if movie['rating'] >= min_rating:
            recommended_list.append(movie['title']) # Append to the list
    return recommended_list


# Use the function to recommend movies rated 9.0 or higher
my_best_movies = recommend_movies(my_movie_dataset, 9.0)
print(f"평점 9.0 이상 추천 영화: {my_best_movies}")

# Reuse the same function to recommend movies rated 9.1 or higher
my_good_movies = recommend_movies(my_movie_dataset, 9.1)
print(f"평점 9.1 이상 추천 영화: {my_good_movies}")
````
