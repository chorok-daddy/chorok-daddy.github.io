## Reuse with a Function

When applying the same decision rule to multiple samples, define the inputs, Threshold, and outputs as the function contract. Handle empty data and nonnumeric inputs separately from valid scores.

<span id="풀이-예제"></span>

## Worked example

Scores=\[45,60,75,90\], and the passing criterion is 60 or higher.

```python
scores=[45,60,75,90]
def pass_flags(values,threshold=60):
    return [v>=threshold for v in values]
print(pass_flags(scores))
```
