## 📦 Module 2: Variables, the Smallest Units of Data {#package-모듈-2-데이터의-가장-작은-단위-변수variables}

- **What Is a Variable?**: A box with a \'name tag\' for storing data (values)
- **Variables in Machine Learning**:
  - A value stored in one cell of a dataset
  - For example, a student\'s \'study hours\' or \'exam score\'
- **Data Types**: The forms of data stored in variables
  - `int`: **Integer** (e.g., 92, -5, 0)
  - `float`: **Floating-point number** (a number with a decimal point, e.g., 10.5, 3.14)
  - `str`: **String** (characters enclosed in quotes, e.g., \"Kim\", \"Pass\")
  - `bool`: **Boolean** (only true or false, e.g., `True`, `False`)

```python
# Store 'one student's' data in variables.

# 1. Store the floating-point (float) value 10.5 in a variable labeled 'study hours'
study_hours = 10.5

# 2. Store the integer (int) value 92 in a variable labeled 'exam score'
exam_score = 92

# Check the stored values
print("공부 시간:", study_hours)
print("시험 점수:", exam_score)

# Check variable data types using type()
print("\n'study_hours' 변수의 자료형:", type(study_hours))
print("'exam_score' 변수의 자료형:", type(exam_score))
```

```python
# Store additional student information

# Store the string (str) value "Kim" in a variable labeled 'student name'
student_name = "Kim"

# Store the Boolean (bool) value True in a variable labeled 'pass status'
is_passed = True

# Check the stored values
print("학생 이름:", student_name)
print("합격 여부:", is_passed)

# Check data types
print("\n'student_name' 변수의 자료형:", type(student_name))
print("'is_passed' 변수의 자료형:", type(is_passed))
```

```python
# Variables can also be used in calculations.
# For example, suppose this student answered a 5-point bonus question correctly.

print("기존 점수:", exam_score)

# Add 5 to the existing score and update it
final_score = exam_score + 5

print("최종 점수:", final_score)
```

## 🧺 Module 3: Data Structures, Containers for Datasets {#basket-모듈-3-데이터셋을-담는-그릇-자료구조}

How can we handle multiple students\' data at once?

We need a larger container, a **data structure**, that can hold multiple values in one variable.

### 1. Lists {#1-리스트-list}

- **Definition**:
  - A data structure that stores multiple values **in order**
  - Delimited by square brackets `[]`
- **Lists in Machine Learning**: Useful for grouping and processing data in order
- **Example**: Lists of all students\' \'study hours\' and \'exam scores\'

```python
# Create lists of 'study hours' and 'exam scores' for multiple students.
study_hours_list = [2.0, 4.5, 6.0, 8.5, 10.0]
exam_scores_list = [65, 78, 85, 95, 98]

print("공부 시간 리스트:", study_hours_list)
print("시험 점수 리스트:", exam_scores_list)

# Lists have an order, so specific values can be accessed by index.
# Python starts counting at 0!
print("\n첫 번째 학생의 공부 시간:", study_hours_list[0])
print("세 번째 학생의 시험 점수:", exam_scores_list[2])

# Use len() to check how many data items a list contains.
print("\n총 학생 수:", len(study_hours_list))
```

```python
# Suppose a new student's data have been added.
# Use .append() to add a value to the end of a list.

print("추가 전 공부 시간 리스트:", study_hours_list)

study_hours_list.append(12.0)
exam_scores_list.append(100)

print("추가 후 공부 시간 리스트:", study_hours_list)
print("추가 후 총 학생 수:", len(study_hours_list))
```

### 2. Dictionaries {#2-딕셔너리-dictionary}

- **Definition**:
  - A data structure that stores \'keys\' and \'values\' as pairs
  - Delimited by braces `{}`
  - Has no order
- **Dictionaries in Machine Learning**: Useful for grouping and processing data according to their relationships
- **Example**: A collection of information containing a student\'s \'name\', \'study hours\', \'exam score\', and \'pass status\'

```python
# Create a dictionary containing all information about one student.
student_A_info = {
    'name': 'Park',
    'study_hours': 8.5,
    'exam_score': 95,
    'is_passed': True
}

print("학생 A의 정보:", student_A_info)

# A dictionary accesses values using Keys rather than positions (indices).
print("\n학생 A의 이름:", student_A_info['name'])
print("학생 A의 시험 점수:", student_A_info['exam_score'])
```

### Final form: Combining lists and dictionaries {#최종-형태-리스트와-딕셔너리의-조합}

- Most Machine Learning data is represented as a **\'list of dictionaries\'** or a **\'dictionary of lists\'**.
- In particular, **\"combining each student\'s information (dictionary) into one list\"** is very common.
  - Like a table in Excel

```python
# Group multiple students' information (dictionaries) in one list to represent the full dataset
full_dataset = [
    { 'name': 'Kim', 'study_hours': 2.0, 'exam_score': 65, 'is_passed': True },
    { 'name': 'Lee', 'study_hours': 4.5, 'exam_score': 78, 'is_passed': True },
    { 'name': 'Park', 'study_hours': 8.5, 'exam_score': 95, 'is_passed': True },
    { 'name': 'Choi', 'study_hours': 1.5, 'exam_score': 55, 'is_passed': False }
]

# Access the first student's ('Kim') information
first_student = full_dataset[0]
print("첫 번째 학생의 정보:", first_student)

# Access only the first student's 'exam score'
first_student_score = full_dataset[0]['exam_score']
print("첫 번째 학생의 시험 점수:", first_student_score)
```
