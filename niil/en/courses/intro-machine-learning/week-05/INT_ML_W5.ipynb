## 🐢 Module 1: Why Are Python Lists Not Enough? {#turtle-모듈-1-왜-파이썬-리스트로는-부족할까}

- **Scenario**
  - Increase the exam scores of two student groups by 5 points each.
- **Limitations of Python lists**
  - Cannot directly perform the simple operation of adding 5 to every number in a list
  - `[80, 90, 70] + 5` → **Error!**
    - A list and a number cannot be added
- **Solution**
  - Use a `for` loop to take each number and add 5,
  - then put the results in a new list.
- This becomes very cumbersome and slow as the data grows!

```python
# Manage scores using a Python list
scores_list = [80, 95, 72, 100, 88]

# Steps to increase every score by 5
new_scores_list = [] # An empty list for the new scores
for score in scores_list:
    # Take each value, add 5, and append it to the new list
    new_scores_list.append(score + 5)

print(f"원본 점수 리스트: {scores_list}")
print(f"5점 올린 후 리스트: {new_scores_list}")
print("\n데이터가 5개만 있어도 이렇게 번거로운데, 100만 개라면...?")
```

## ✨ Module 2: Introducing NumPy Arrays {#sparkles-모듈-2-numpy-배열array의-등장}

- **What Is NumPy?**
  - Short for Numerical Python
  - A library for processing large amounts of numerical data quickly and efficiently
- **NumPy Array (ndarray)**
  - A multidimensional array (list) used in NumPy
  - Constraint: **Stores only data of the same type**
    - Numbers together or text together
    - This constraint enables very fast operations!
  - **Vectorized Operations**
    - Apply an operation to all array elements at once without a loop.
    - The most important feature!

```python
import numpy as np # Conventionally imported with the alias np.

# Convert a Python list into a NumPy array
scores_array = np.array([80, 95, 72, 100, 88])

print(f"NumPy 배열: {scores_array}")
print(f"자료형: {type(scores_array)}")
```

```python
# Use a NumPy array to increase every score by 5
# No loop is needed!

new_scores_array = scores_array + 5

print(f"원본 점수 배열: {scores_array}")
print(f"5점 올린 후 배열: {new_scores_array}")

# All arithmetic operations are available, not just addition.
print(f"\n모든 점수 10% 인상: {scores_array * 1.1}")
print(f"모든 점수 2로 나누기: {scores_array / 2}")
```

## 🚀 Module 3: Essential NumPy Features for Machine Learning {#rocket-모듈-3-machine-learning을-위한-numpy-핵심-기능}

### 1. Array Shape {#1-배열의-형태-shape}

- **What is shape?**
  - Information about how many rows and columns an array has.
  - A fundamental and important part of understanding data structure in Machine Learning.
- **1-dimensional array**: Vector
  - e.g., scores of 5 students
- **2-dimensional array**: Matrix
  - e.g., scores of 5 students in 2 subjects, Korean and English
- **3-dimensional array**: Tensor
  - e.g., a color image with pixels indexed by (width, height, color)

```python
# 1-dimensional array (Vector)
# [상자1, 상자2, 상자3, 상자4, 상자5, 상자6] -> 6 items in one row
array_1d = np.array([1, 2, 3, 4, 5, 6])
print(f"1D 배열: {array_1d}")
print(f"1D 배열의 Shape: {array_1d.shape}") # (6,) -> One row containing 6 items

print("-" * 20)

# 2-dimensional array (Matrix)
# [[국어1, 영어1], -> First line (row)
# [국어2, 영어2], -> Second line (row)
# [국어3, 영어3]] -> Third line (row)
array_2d = np.array([
    [90, 85], # Student 1
    [75, 92], # Student 2
    [88, 80]  # Student 3
])
print(f"2D 배열:\n{array_2d}")
print(f"2D 배열의 Shape: {array_2d.shape}")
# (3, 2) -> 3 rows, 2 columns
```

### 2. Changing Array Shape (Reshape) {#2-배열의-형태-바꾸기-reshape}

- `reshape()`
  - Use this to keep the data contents and total count unchanged
  - while changing only the array shape.
- **Using -1**
  - A wildcard meaning **\"Calculate the remaining dimension automatically.\"**
  - `array.reshape(2, -1)`
    - \"There are 6 elements; arrange them into 2 rows. Calculate the column count automatically.\"
    - → Reshaped to (2, 3)
  - `array.reshape(-1, 3)`
    - \"There are 6 elements; arrange them into 3 columns. Calculate the row count automatically.\"
    - → Reshaped to (2, 3)
  - **Note**: An error occurs if the total element count is not divisible by the specified dimension.
    - e.g., 6 elements cannot form 4 rows

```python
# 1-dimensional array (6 items)
array_1d = np.array([1, 2, 3, 4, 5, 6])
print(f"원본 1D 배열: {array_1d}")
print(f"원본 1D 배열 Shape: {array_1d.shape}")

# Reshape into a 2-dimensional array with 2 rows and 3 columns
# [1,2,3,4,5,6] -> [[1,2,3],
#                   [4,5,6]]
array_reshaped = array_1d.reshape(2, 3)
print(f"\n(2, 3)으로 Reshape:\n{array_reshaped}")
print(f"Shape: {array_reshaped.shape}")

# Use -1 to calculate the number of columns automatically
# "Reshape 6 elements into 3 rows. Determine the columns automatically!" -> Requires 2 columns
array_reshaped_auto = array_1d.reshape(3, -1)
print(f"\n(3, -1)으로 Reshape:\n{array_reshaped_auto}")
print(f"Shape: {array_reshaped_auto.shape}")
```

### 3. Operations Along an Axis {#3-축axis을-따른-연산}

- **What is an axis?**
  - Specifies the **direction of an operation** in a multidimensional array.
- **Visualize It (in 2 Dimensions)**
  - **`axis=0`**: **Vertical direction (↓)**
    - Operate on each **column**.
    - Imagine compressing multiple rows down into one row.
  - **`axis=1`**: **Horizontal direction (→)**
    - Operate on each **row**.
    - Imagine compressing multiple columns into one column.
- If `axis` is not specified, the operation applies to all elements of the array.

```python
# Korean and English scores for 3 students
scores_2d = np.array([
    [90, 85], # Student 1
    [75, 92], # Student 2
    [88, 80]  # Student 3
])
print(f"원본 점수 배열 (Shape: {scores_2d.shape}):\n{scores_2d}")

# 1. Overall sum / mean
print(f"\n전체 총점: {scores_2d.sum()}")
print(f"전체 평균: {scores_2d.mean():.2f}")

# 2. axis=0 (vertical direction ↓) operations
# Sum/mean of each 'column' -> Statistics by subject
print(f"\n--- axis=0 (과목별 통계) ---")

print(f"과목별 총점: {scores_2d.sum(axis=0)}")
# [90+75+88, 85+92+80]

print(f"과목별 평균: {scores_2d.mean(axis=0)}")
# [avg(90,75,88), avg(85,92,80)]

print(f"과목별 최고점: {scores_2d.max(axis=0)}")
# [max(90,75,88), max(85,92,80)]

# 3. axis=1 (horizontal direction →) operations
# Sum/mean of each 'row' -> Statistics by student
print(f"\n--- axis=1 (학생별 통계) ---")

print(f"학생별 총점: {scores_2d.sum(axis=1)}")
# [90+85, 75+92, 88+80]

print(f"학생별 평균: {scores_2d.mean(axis=1)}")
# [avg(90,85), avg(75,92), avg(88,80)]

print(f"학생별 최고점: {scores_2d.max(axis=1)}")
# [max(90,85), max(75,92), max(88,80)]
```

### 4. Useful Array Creation Functions {#4-유용한-배열-생성-함수}

- `np.arange(n)`: Creates an array containing numbers from 0 to n-1
- `np.zeros(shape)`: Create an array filled with 0 (useful for initialization).
- `np.ones(shape)`: Create an array filled with 1

```python
# An array containing the numbers 0 through 9
range_array = np.arange(10)
print(f"np.arange(10):\n{range_array}")

# An array filled with 0, with shape (2, 3)
zeros_array = np.zeros((2, 3))
print(f"\nnp.zeros((2, 3)):\n{zeros_array}")

# An array filled with 1, with shape (3, 2)
ones_array = np.ones((3, 2))
print(f"\nnp.ones((3, 2)):\n{ones_array}")
```
