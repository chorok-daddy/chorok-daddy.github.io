## Topics covered {#이번에-배울-내용}

Choose model complexity by checking whether a tree that fits the training data well also performs well on new data.

## Tree Questions and Complexity {#트리의-질문과-복잡도}

A Decision Tree splits data using feature values. Greater depth can fit detailed differences in training data but increase errors on new data.

## Comparing Models under the Same Conditions {#같은-조건의-모델-비교}

Random Forest combines results from multiple trees. Compare it with a single tree using the same split, preprocessing, and metrics to interpret differences in complexity. Examine both accuracy and errors by class.

## Worked example {#풀이-예제}

Split the Iris data with test_size=.3, random_state=7, stratify=y. Use depths 1,3,None.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
X,y=load_iris(return_X_y=True)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=7,stratify=y)
model=DecisionTreeClassifier(max_depth=3,random_state=7).fit(Xtr,ytr)
print(model.score(Xte,yte))
```
