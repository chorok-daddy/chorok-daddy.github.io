## Select the required values from JSON

<span id="시작-조건"></span>

### Starting conditions

The sensor response contains device, timestamp, and temperature_c. The second row lacks a temperature, and the third has a numeric string.

<span id="진행-방법"></span>

### Procedure

1.  Read the provided JSON and check each row's fields and data types.
2.  Distinguish successful collection from values suitable for analysis.

<span id="남길-결과"></span>

### Results to retain

- A table of valid, missing-value, and format conversion cases.

```python
import json
payload='[{"device":"a","timestamp":0,"temperature_c":21.2},{"device":"a","timestamp":1},{"device":"b","timestamp":0,"temperature_c":"22.1"}]'
rows=json.loads(payload)
def clean(row):
    if 'temperature_c' not in row:return None,'missing temperature'
    try:return dict(row,temperature_c=float(row['temperature_c'])),'ok'
    except (ValueError,TypeError):return None,'invalid temperature'
for row in rows:print(clean(row))
```
