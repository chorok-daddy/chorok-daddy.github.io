# External data collection and quality checks · Learning {#외부-데이터-수집과-품질-점검--학습}

## Topics covered {#이번에-배울-내용}

Check both the HTML structure and extraction rules so that missing fields on a page do not cause other records to be discarded.

## Overall workflow {#전체-흐름}

JSON and HTML responses → field and unit checks → valid and rejected data → CSV storage

## Successful collection and value quality {#수집-성공과-값의-품질}

Receiving JSON does not make it ready for analysis. Check field presence, data types, physical ranges, and units in order. Replacing missing values with actual 0 values can distort means and alarm thresholds.

## Boundaries to preserve when the input format changes {#입력-형식이-바뀌어도-지킬-경계}

Separate collection, cleaning, and storage into functions to limit the impact of API field or HTML structure changes. Record reasons for rejected rows, and check that numbers become strings during a CSV round trip.

## Worked example {#풀이-예제}

The sensor response contains device, timestamp, and temperature_c. The second row lacks a temperature, and the third has a numeric string.

```python
import json
payload='[{"device":"a","timestamp":0,"temperature_c":21.2},{"device":"a","timestamp":1},{"device":"b","timestamp":0,"temperature_c":"22.1"}]'
rows=json.loads(payload)
def clean(row):
    if 'temperature_c' not in row:return None,'missing temperature'
    try:return dict(row,temperature_c=float(row['temperature_c'])),'ok'
    except (ValueError,TypeError):return None,'invalid temperature'
for row in rows:print(clean(row))
```
