## Topics covered

Choosing a resolution and capture interval requires understanding the trade-offs among storage, field of view, and frame rate.

<span id="전체-흐름"></span>

## Overall workflow

Frame array → ROI and resolution → timestamps → memory and actual FPS

<span id="한-frame의-비용"></span>

## Cost of one frame

Uncompressed storage is height × width × number of channels × bytes per channel. Doubling the resolution along both axes quadruples the pixel count. Using only the required region as an ROI lets you adjust both observed information and processing cost.

<span id="요청한-fps와-실제-기록"></span>

## Requested FPS and actual records

Exposure, transmission, conversion, and storage take time. Calculate actual FPS from timestamp intervals. Counting frames alone or recording only requested settings can miss interruptions and delays.

<span id="풀이-예제"></span>

## Worked example

Use the 120×160×3 uint8 RGB image generated by the provided NumPy code.

```python
import numpy as np
import matplotlib.pyplot as plt
x,y=np.meshgrid(np.linspace(0,255,160),np.linspace(0,255,120))
frame=np.stack([x,y,255-x],axis=-1).astype(np.uint8)
small=frame[::2,::2]
print(frame.shape,frame.nbytes,small.shape,small.nbytes)
```
