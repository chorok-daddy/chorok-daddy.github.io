## Topics covered {#이번에-배울-내용}

Misinterpreting array axes or time intervals can give a physically incorrect meaning even when the calculation runs.

## Overall workflow {#전체-흐름}

Sensor and request → format and unit checks → storage and state changes → query and response checks

## Assign meaning to array axes {#배열-축에-의미-붙이기}

If rows represent measurement times and columns represent sensor channels, `axis=0` aggregation summarizes each channel. A correct Shape does not ensure the intended calculation. Choose elementwise or matrix multiplication according to its physical meaning.

## Time and noise {#시간과-잡음}

To interpret differences as velocity, divide by the measurement time interval. A moving average reduces noise but blurs abrupt changes, and the sample count may differ at boundaries. Compare before and after filtering over the same time interval.

## Worked example {#풀이-예제}

$A=\begin{bmatrix}1&2&3\\4&5&6\end{bmatrix}$, $B=\begin{bmatrix}2&0\\1&3\\0&1\end{bmatrix}$.

```python
import numpy as np
A=np.array([[1,2,3],[4,5,6]])
B=np.array([[2,0],[1,3],[0,1]])
print(A@B)
```
