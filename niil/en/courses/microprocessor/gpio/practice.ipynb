### Results to retain

```python
def debounce(times,min_gap):
    accepted=[]
    for t in times:
        if not accepted or t-accepted[-1]>=min_gap:accepted.append(t)
    return accepted
print(debounce(rising,20))
# TODO: Compare 5, 20, and 50 ms.
```
