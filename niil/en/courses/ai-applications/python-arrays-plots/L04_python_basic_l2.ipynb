### 2.1 Comments

- Single-line comment: A statement beginning with `#` is a comment.

```python
# helloworld
print('hello world')
```

- Comment block: A block enclosed in `'''` (three single quotes) or `"""` (three double quotes) acts as a comment.

```python
'''
Function name: helloworld()
This function prints 'hello world'
'''
def helloworld():
    print('hello world')

helloworld()
```

<span id="들여쓰기와-코드블럭"></span>

### 2.2 Indentation and code blocks

- Indentation plays a very important role in Python, so use it carefully.

- Incorrect indentation causes an error. `Indentation Error`

- Indentation is determined by the number of spaces or tabs; using tabs is recommended for readability and consistency.

- **Role 1**: Indicates the scope of a loop or control statement.

```python
print('start')
for i in range(10):
    if i>5:
        print('i = %d > 5' % i)
        # The if block ends here.
    elif i<5:
        print('i = %d < 5' % i)
    else:
        print('i = %d = 5' % i)
    # The for block ends here.
print('end')
```

- **Role 2**: Indicates the scope of a function or class definition.

```python
def function1():
    print('  function start')
    print('  function end')
# The function body ends here.

print('script start')
function1()
print('script end')
```

### 2.3 Input and output

- Input `input()`
  - This function accepts keyboard input through the screen and stores it in a variable.

```python
a = input()
print(a)
```

- Output `print()`
  - This function displays the desired content on the screen.
  - Write the expression to display between `' '`.
  - For content that varies with a variable, use `%d` (integer), `%f` (real number), or `%s` (string), according to its type.
    - `%.1f` (display a real number to one decimal place)
  - Put `%` to the right of `' '`, then list the variables to substitute into the expression in order inside `( )`.

```python
i = 4
j = 4.5
k = 'abc'
print('i=%d j=%.1f k=%s' % (i,j,k))
```
