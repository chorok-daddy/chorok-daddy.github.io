## 2. Operations and verification {#2-연산과-검증}

Compute and compare A\*B, which multiplies matching elements, and [A@B](mailto:A@B를){.email}, which sums row-column products. Then compute A\'s Inverse and check whether [A@Inverse](mailto:A@Inverse가){.email} is the Identity.

### A. Baseline example · Read and run first {#a-기준-예제--먼저-읽고-실행}

````python
A=np.array([[2.,1.],[1.,3.]])
B=np.array([[1.,0.],[2.,1.]])
print('elementwise:',A*B,'matrix:',A@B)
print('inverse:',np.linalg.inv(A))
assert np.allclose(A@np.linalg.inv(A),np.eye(2))
values=np.array([3.,7.,5.]);print('max:',values.max(),'index:',values.argmax())
````

## 3. Plot the results {#3-결과를-그래프로}

Compute the maximum\'s position and mark it on the plot.

### A. Baseline example · Read and run first {#a-기준-예제--먼저-읽고-실행}

````python
x=np.linspace(0,2*np.pi,100)
y=np.sin(x)
fig,ax=plt.subplots();ax.plot(x,y,label='sin(x)');i=y.argmax()
ax.annotate('sample maximum',(x[i],y[i]),xytext=(2,0.5),arrowprops={'arrowstyle':'->'})
ax.set(xlabel='x (rad)',ylabel='value',title='Sampled sine');ax.grid();ax.legend();plt.show()
````
