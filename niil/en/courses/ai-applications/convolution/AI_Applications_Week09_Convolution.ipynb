## 1. Inspect layer shapes with a simple CNN {#1-간단한-cnn으로-각-층의-shape-확인하기}

First run a model with one Conv Layer, then build a CNN with three Conv Layers yourself.

### A. Baseline example · Read and run first {#a-기준-예제--먼저-읽고-실행}

````python
import sys, numpy as np, matplotlib.pyplot as plt
import torch
from torch import nn
import torch.nn.functional as F
torch.manual_seed(42)
torch.set_num_threads(2)
print('Python:', sys.version.split()[0], '| PyTorch:', torch.__version__)
small=nn.Sequential(nn.Conv2d(1,4,3,padding=1),nn.ReLU(),nn.MaxPool2d(2),nn.Flatten(1),nn.Linear(4*14*14,10))
x=torch.randn(8,1,28,28)
for layer in small:
    x=layer(x);print(type(layer).__name__,tuple(x.shape))
assert x.shape==(8,10)
````
