## From arrays to images

<span id="시작-조건"></span>

### Starting conditions

Draw a blue rectangle in a 120×160 BGR image, and convert it to RGB, grayscale, and HSV.

<span id="진행-방법"></span>

### Procedure

1.  Check the original array's shape, dtype, and range.
2.  Display the same image correctly in Matplotlib and compare with a deliberately altered channel order.

<span id="남길-결과"></span>

### Results to retain

- Comparison images for each representation and an array summary.

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
image=np.full((120,160,3),220,dtype=np.uint8);image[30:90,50:110]=(255,30,30)
rgb=cv.cvtColor(image,cv.COLOR_BGR2RGB);hsv=cv.cvtColor(image,cv.COLOR_BGR2HSV)
fig,axes=plt.subplots(1,2);axes[0].imshow(image);axes[0].set_title('BGR as RGB');axes[1].imshow(rgb);axes[1].set_title('Correct RGB')
print(image.shape,image.dtype,hsv[60,80])
```

## Pass coordinates to the next stage

<span id="시작-조건-2"></span>

### Starting conditions

Calculate the center of the region detected in the mask in pixel coordinates.

<span id="진행-방법-2"></span>

### Procedure

### Results to retain
