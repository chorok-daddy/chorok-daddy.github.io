# Notebook Basics · Study

<span id="이번에-배울-내용"></span>

## Topics covered

Code alone makes it easy to forget inputs and the reasoning behind decisions. Turn a small experiment into a record you can revisit and check.

<span id="전체-흐름"></span>

## Overall workflow

Inputs and units → code/output → explanations/figures → rerun from the beginning

<span id="계산과-설명을-한-기록에-두기"></span>

## Keep calculations and explanations in one record

Code records calculations; Markdown records input meanings, units, and decision criteria. Instead of pasting only the output numbers, describe the inputs directly above them. Label conditions, axes, and units in tables and plots.

<span id="다시-실행할-수-있는-기록"></span>

## A reproducible record

Notebook cells can be run in different orders, making it easy to depend on variables absent from the visible code. After finishing, restart the kernel and run from the top. Check that the modified conditions and latest output refer to the same experiment.

<span id="풀이-예제"></span>

## Worked example

If distances of 1.2, 1.0, and 0.9 m are recorded at times 0, 1, and 2 seconds, the interval rates of distance change are −0.2 and −0.1 m/s. State inputs and units before the code, and explain the negative signs beside the results.
