## From input events to state

A key press sets a velocity command, and a release clears it. Adding distance for every received event can make movement depend on the number of repeated inputs and processing speed. Maintain state and integrate position over time intervals.

<span id="명령의-유효-시간"></span>

## Command validity duration

Continuing at the last velocity when input stops is dangerous. Track the last reception time and use a timeout to replace stale commands with a stop. Test normal termination, communication gaps, and repeated inputs separately.
