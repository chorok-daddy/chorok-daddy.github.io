## Gain and closed-loop poles

<span id="시작-조건"></span>

### Starting conditions

$L(s)=\frac{s+7}{s(s+5)(s+15)(s+20)}$. Use $K\in\{100,500,1000\}$.

<span id="진행-방법"></span>

### Procedure

1.  Plot the root locus and mark the poles of `feedback(K*L,1)` for each K using distinct symbols.
2.  Tabulate the real parts of the poles and stability for each case.
3.  Predict pole movements and response changes relative to the baseline condition $K=500$.

<span id="남길-결과"></span>

### Results to retain

- 1 root-locus plot and a pole table for each K.

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ctl
s = ctl.tf('s')
L=(s+7)/(s*(s+5)*(s+15)*(s+20))
ctl.root_locus_plot(L)
for k in [100,500,1000]: print(k,ctl.poles(ctl.feedback(k*L,1)))
```

## Testing predictions with time responses

<span id="시작-조건-1"></span>

### Starting conditions

Apply a unit step over $0\le t\le 8$ s to the same model and gain.

<span id="진행-방법-1"></span>

### Procedure

1.  Calculate overshoot and settling time only for stable cases.
2.  Explain the relationship between the slowest pole pair and oscillation and decay in the time response.
3.  $K=500$: add one value above or below this and test your prediction.

<span id="남길-결과-1"></span>

### Results to retain

- A response plot and performance table including all 4 conditions.

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ctl
s = ctl.tf('s')
L=(s+7)/(s*(s+5)*(s+15)*(s+20))
t=np.linspace(0,8,1601)
T=ctl.feedback(500*L,1)
t,y=ctl.step_response(T,t)
plt.plot(t,y);plt.xlabel('Time [s]');plt.grid()
# TODO: Add the remaining gains.
```
