## Laplace transform and inverse transform

<span id="시작-조건-1"></span>

### Starting conditions

Transform $f(t)=e^{-2t}$ for t\>0, then take its inverse transform.

<span id="진행-방법-1"></span>

### Procedure

1.  Use sym.laplace_transform and inverse_laplace_transform to obtain the transform and inverse transform.
2.  Compare the original and recovered function values at $t\in\{.5,1,2\}$.
3.  Explain how the pole −2 of the transformed expression relates to the time-domain decay term $e^{-2t}$.

<span id="남길-결과-1"></span>

### Results to retain

- A short table of F(s), the recovered function, and numerical differences.

```python
import sympy as sym
t=sym.symbols('t',positive=True);s=sym.symbols('s')
f=sym.exp(-2*t)
F=sym.laplace_transform(f,t,s,noconds=True)
restored=sym.inverse_laplace_transform(F,s,t)
print(F,restored)
# TODO: Check the differences at the three specified times.
```
