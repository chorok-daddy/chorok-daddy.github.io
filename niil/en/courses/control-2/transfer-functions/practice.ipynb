## Comparing series, parallel, and feedback connections

<span id="시작-조건-1"></span>

### Starting conditions

$H_1(s)=\frac2s$, $H_2(s)=\frac3{4s+1}$. Use negative unity feedback for $H_1H_2$.

<span id="진행-방법-1"></span>

### Procedure

1.  Derive expressions for $H_s=H_1H_2$, $H_p=H_1+H_2$, and $T=\frac{H_s}{1+H_s}$.
2.  Use ctl.series, ctl.parallel, ctl.feedback to build the same models, and compare with the directly derived expressions at ω=\[0.1,1,10\] rad/s.
3.  Plot the unit-step response of T over $0\le t\le 50$ s. Distinguish the poles at the origin of Hs and Hp from the meaning of DC gain.

<span id="남길-결과-1"></span>

### Results to retain

- 1 coefficient table for the 3 interconnections and 1 step plot of T.
- Record the error as the absolute difference between complex responses at the same frequency.

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ctl
s = ctl.tf('s')
H1=ctl.tf([2],[1,0]); H2=ctl.tf([3],[4,1])
Hs=ctl.series(H1,H2); Hp=ctl.parallel(H1,H2); T=ctl.feedback(Hs,1)
print(Hs, Hp, T)
t=np.linspace(0,50,2501)
t,y=ctl.step_response(T,t)
plt.plot(t,y); plt.xlabel('Time [s]'); plt.ylabel('Output'); plt.grid()
# TODO: Compare frequency responses with the directly calculated models.
```
