## Where the three techniques differ {#세-기법의-다른-위치}

Double changes next-action selection and value evaluation; Dueling changes the Q representation; PER changes experience sampling. They do not all address the same problem.

## Representation and distribution {#표현과-분포}

Dueling can combine streams using $Q=V+A-\operatorname{mean}(A)$. PER samples large-error experiences more frequently while accounting for sampling bias and weights. It does not create states outside the experience coverage.

## Worked example {#풀이-예제}

- $Q_{\mathrm{online}}=[2,4,3]$, $Q_{\mathrm{target}}=[5,1,2]$.
- $r=1$, $\gamma=.95$.
- Dueling: $V=2$, $A=[1,2,6]$.
- TD Error: $[0.1,1,5]$.

```python
import numpy as np
online=np.array([2.,4.,3.]);target=np.array([5.,1.,2.])
print('DQN',1+.95*target.max(),'Double',1+.95*target[online.argmax()])
advantage=np.array([1.,2.,6.]);print('Dueling Q',2+advantage-advantage.mean())
error=np.array([.1,1.,5.])
for alpha in [0,.6]:
    probability=(error+1e-5)**alpha;probability/=probability.sum();print(alpha,probability)
```
