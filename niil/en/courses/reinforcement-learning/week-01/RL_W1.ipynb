## \[What is Reinforcement Learning?\] {#reinforcement-learning이란}

### 1. Definition {#1-정의}

- A field of machine learning in which an agent interacts with an environment to learn a policy that maximizes reward

### 2. Main components {#2-주요-구성-요소}

- **Agent**
  - The entity that selects actions according to a policy
- **Environment**
  - The external world that receives the agent\'s action and returns the next state and reward
- **State**
  - Information describing the environment\'s current situation
    - Example: Cart position and pole angle in CartPole
- **Action**
  - A choice the agent can make in the environment
    - Example: Move the cart left or right.
- **Reward**
  - A real-valued signal received from the environment as a result of an action
  - Goal: Improve the policy to maximize this reward.
- **Policy**
  - Definition: A strategy for deciding which action to take in a given state
  - Stochastic policy: π(a\|s) is the probability of choosing action a in state s.

### 3. The objective of reinforcement learning {#3-reinforcement-learning의-목표}

- Definition: find a policy that maximizes expected cumulative reward (return)
- Training process
  - Initialize the environment and receive the state
  - Take an action
  - Obtain the next state and reward
  - Construct and repeat episodes
  - Gradually improve the policy

### 4. Policy {#4-정책policy}

- Definition: A strategy for deciding which action to take in a given state
- Feature: takes a state and outputs a probability distribution over actions
- Stochastic policy: π(a\|s) is the probability of choosing action a in state s

```python
env = gym.make('CartPole-v1')
```
