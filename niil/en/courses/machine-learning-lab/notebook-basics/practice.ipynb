## Comparing measurements and alarm labels {#측정값과-alarm-label-비교}

### Starting conditions {#시작-조건}

The times are \[0,5,10,15,20,25\] min and temperatures are \[24,25,29,23,30,26\] °C. Trigger an alarm when the temperature exceeds the threshold.

### Procedure {#진행-방법}

1.  Run the learning example and compare the alarm times with the temperatures.
2.  Change the threshold from 28°C to 26°C and find the additional alarm times.
3.  Mark the threshold line and alarm points on the same plot, and explain the effect of changing the threshold.

### Results to retain {#남길-결과}

- Alarm times and counts for each threshold, plus a plot with axes and units.

```python
import numpy as np
import matplotlib.pyplot as plt
minutes=np.array([0,5,10,15,20,25])
temperature=np.array([24.,25.,29.,23.,30.,26.])
threshold=28.
alarm=temperature>threshold
print('alarm times [min]:',minutes[alarm])
print('alarm count:',alarm.sum())
plt.plot(minutes,temperature,'o-',label='measurement')
plt.axhline(threshold,color='gray',linestyle='--',label='threshold')
plt.scatter(minutes[alarm],temperature[alarm],color='red',label='alarm')
plt.xlabel('Time [min]');plt.ylabel('Temperature [deg C]');plt.legend();plt.grid()
# TODO: Compare results using the 26°C criterion.
```

## Checking whether calibration reduces error {#보정식이-오차를-줄이는지-확인하기}

### Starting conditions {#시작-조건}

The sensor inputs are $x=\begin{bmatrix}20&22&24&26\end{bmatrix}$ °C, and the reference temperatures are $y=\begin{bmatrix}22.5&24.7&26.8&29.1\end{bmatrix}$ °C. The calibration equation is y_hat=1.1\*x+0.5.

### Procedure {#진행-방법}

1.  Calculate predictions using the input unchanged and using the calibration equation.
2.  Calculate the difference between each prediction and its reference value, and the mean absolute error.
3.  Also compare errors from the two methods for a new input of 28°C and a reference temperature of 31.4°C.

### Results to retain {#남길-결과}

- A table with inputs, reference values, two predictions, and errors side by side, plus a comparison using a new measurement.

```python
x=np.array([20.,22.,24.,26.]);reference=np.array([22.5,24.7,26.8,29.1])
corrected=1.1*x+.5
print(np.column_stack([x,reference,corrected]))
print('raw MAE:',np.mean(np.abs(x-reference)))
print('corrected MAE:',np.mean(np.abs(corrected-reference)))
# TODO: Compute predictions for the new inputs and both errors.
```

## Identifying and recording invalid measurements {#잘못된-측정값을-구분하고-기록하기}

### Starting conditions {#시작-조건}

Use the records \[24,None,29,100,26,28\] °C at the same times. The sensor\'s valid temperature range is 15--40°C, and None indicates a failed measurement.

### Procedure {#진행-방법}

1.  Distinguish failed measurements from out-of-range values and record them with their original times.
2.  Use valid values to calculate the mean and alarms above 28°C. Compare with the results when the condition is changed to ≥.
3.  Retain missing times on the plot and indicate that no measurement is available.

### Results to retain {#남길-결과}

- A table of input status, the mean and alarm count for valid measurements, and a plot that distinguishes missing measurements.

```python
records=[24,None,29,100,26,28]
values=np.array([np.nan if v is None else float(v) for v in records])
valid=np.isfinite(values)&(values>=15)&(values<=40)
print('valid times:',minutes[valid])
print('valid mean:',values[valid].mean())
# TODO: Compare the input conditions and two alarm conditions, and plot with invalid points marked as np.nan.
```
