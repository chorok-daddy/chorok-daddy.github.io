### 4.1 Basic mathematical operations

- Variables
  - A variable is a name identifying a storage location for data. It is called a variable because the stored data can change.
  - For readability, choose names that reflect the data type or purpose.
    - (Reading other people's source code reveals some naming conventions.)

````python
# Good variable examples
pos_x = 10 
num_values = 5
_event_flag = False

# Bad variable examples
abc = 4
a1 = 50
xxxx = 3.4
temp1 = 4
temp2 = 5
````

- Data types
  - These include `integer`, `floating-point number`, `bool` (True/False), and `string`.
  - Python manages them automatically, so you generally do not need to handle them explicitly.
    - However, be careful: a data type can change against your intention.

````python
# Integer
number = 5

# Floating-point number
value = 2.5

# bool
is_number = True

# String (string)
name = 'Ivan'

# When an integer becomes a floating-point number
number = number / 2
print('number =', number)
````

- Basic operators
  - Arithmetic
    - Addition `+`, subtraction `-`, multiplication `*`, division `/`)
  - Other operations
    - Quotient `//`, remainder `%`, exponentiation `**`
  - Logical operations
    - Greater than `>`, greater than or equal to `>=`, less than `<`, less than or equal to `<=`, equal to `==`, not equal to `!=`, `and`, `or`

````python
a = 10
b = 2.5
print(a+b, a-b, a*b, a/b, a//b, a%b, a**b)
print(a>b, a>=b, a<b, a<=b, a==b, a!=b, a>b or a<=b, a>b and a<=b)
````

- Common mathematical functions (using NumPy)
  - Sine `sin`, cosine `cos`, exponential `exp`, round `round`, ceiling `ceil`, floor `floor`

````python
# We will cover use of the numpy library in more detail later.
import numpy as np
angle = np.pi/4
print(angle)
print(np.sin(angle))
print(np.cos(angle))
print(np.exp(angle))
print(np.round(angle))
print(np.ceil(angle))
print(np.floor(angle))
````

### 4.2 Simple loops (for)

- The `for` statement is a built-in feature commonly used to `repeat` a specified number of times.
- Format

<span id="cb1"></span>

``` python
    for variable in items:
        statement
        ...
```

    + Each list element is assigned to the variable in order during iteration, and the loop ends after all elements have been processed. 
    + The list can contain any data type.

````python
for i in [4, 'text', 1.5 , True, {'name':'Charles', 'age':15}]:
    print(i)
````

- However, the following form is more common.

````python
for i in range(5):
    print(i)
````

- The `continue` statement can skip **one** iteration. (The `if` statement is explained in more detail at the end of this section.)

````python
for i in range(5):
    if i==3:
        continue
    print(i)
````

- The `break` statement can **completely** end a loop. (`break` exits only one loop at a time.)

````python
for i in range(5):
    if i==3:
        break
    for j in range(5):
        if j==2:
            break
        print(i, j)
````

- (Simple example) Multiplication table (Print only tables 2–4 to keep the output short.)

````python
for i in range(2,5):
    for j in range(1,10):
        print('%d * %d = %d' % (i, j, i*j))
    print('----------')
````

### 4.3 Conditional loops (while)

- The `while` statement is used to `repeat` only when a `condition` is satisfied.
- Format

<span id="cb3"></span>

``` python
    while(condition):
        statement
        ...
```

    + The condition has type bool and can be expressed as `True` or `False`.
        - Conditions can be combined as (condition `and` condition), (condition `or` condition).

- When the condition is `True`, the statements below execute, then the condition is checked again.
- When the condition is `False`, the loop stops and execution continues with the next statement.
- When the condition is always `True`, the loop has no way to stop and is often called an **infinite loop**.
  - Unless an infinite loop is intentional, carefully check whether some conditions could cause one.
  - `continue` and `break` work the same way as in a `for` statement, and `break` can exit an infinite loop. (See the example below.)

````python
i = [1,2,3,4,-1,5,6,7,8]
j = []
count=0
while(i[count]>0):
    j.append(i[count])
    count = count + 1
print(j)
````

- (Simple example) Multiplication table (Accept input and print the requested multiplication table.)

````python
# Provide information about how the program works.
print('원하는 단수를 입력하시고, 종료하려면 -1을 입력하세요.')

# Deliberately use an infinite loop.
while(True):
    
    # Assume the user enters an integer.
    i = int(input())
    
    # Exit the loop when -1 is entered.
    if i==-1:
        break
    
    # Output
    for j in range(1,10):
        print('%d x %d = %d' % (i, j, i*j))
````

### 4.4 More complex calculations (if)

- The `if` statement performs the desired action when a `condition` is satisfied. (Unlike `while`, it does not repeat.)
- Format

<span id="cb5"></span>

``` python
    if (condition):
        statement
        ...
    elif (condition):
        statement
        ...
    else:
        statement
        ...
```

- The statements below run when the condition is `True`.
- If another condition needs to be checked when the condition is `False`, use the `elif` statement.
- Place any statements to run when the condition is `False` under the `else` statement.
- Examples of `if` statements are omitted because many have already been shown above.
