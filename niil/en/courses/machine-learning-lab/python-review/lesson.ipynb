## Worked example

The 3-feature data are rows=\[\[1,2,3\],\[2,None,4\],\[3,4,5\],\[4,5,6\]\]. Keep rows containing None in a separate list.

```python
import numpy as np
import matplotlib.pyplot as plt
rows=[[1,2,3],[2,None,4],[3,4,5],[4,5,6]]
complete=[row for row in rows if None not in row]
X=np.asarray(complete,dtype=float)
print(X.shape,X.mean(axis=0))
```
