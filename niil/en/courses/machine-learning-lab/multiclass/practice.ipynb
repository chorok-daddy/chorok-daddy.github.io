## Comparing multiclass logits and predictions

<span id="시작-조건"></span>

### Starting conditions

Generate 6 samples per class, 18 in total, around the provided centers (-2,0),(2,0),(0,2). Use `seed=7`.

<span id="진행-방법"></span>

### Procedure

1.  Check the input and label shapes for Linear(2,3) and CrossEntropy.
2.  Record the 3 logits, 3 probabilities, and prediction for each of the first 6 samples.
3.  Plot a scatter plot of the data and a loss curve.

<span id="남길-결과"></span>

### Results to retain

- A 6-row output table and plots of the data and loss.

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch import nn
rng=np.random.default_rng(7)
centers=np.array([[-2.,0.],[2.,0.],[0.,2.]])
X_np=np.concatenate([c+rng.normal(0,.25,(6,2)) for c in centers]);y_np=np.repeat(np.arange(3),6)
X=torch.tensor(X_np,dtype=torch.float32);y=torch.tensor(y_np,dtype=torch.long)
def fit_multi(labels=y):
    torch.manual_seed(7);m=nn.Linear(2,3);opt=torch.optim.SGD(m.parameters(),lr=.1);history=[]
    for _ in range(200):
        loss=nn.functional.cross_entropy(m(X),labels);opt.zero_grad();loss.backward();opt.step();history.append(loss.item())
    return m,history
model,history=fit_multi();logits=model(X).detach();prob=torch.softmax(logits,dim=1);pred=prob.argmax(dim=1)
print(prob[:3],pred[:3])
```
