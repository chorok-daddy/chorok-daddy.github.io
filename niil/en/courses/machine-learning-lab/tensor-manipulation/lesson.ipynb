## Batch and feature axes

With three samples and four features, the shape is (3,4). cat concatenates along an existing axis, while stack creates a new axis. Track the meaning of samples and features as axes are added or removed.

<span id="계산-가능과-올바른-계산"></span>

## A valid operation versus a correct calculation

Broadcasting expands arrays of different sizes for operations, but it can also act along unintended axes. Writing the expected shape before multiplication and checking the results with a small numerical example helps reveal silent errors.

<span id="풀이-예제"></span>

## Worked example

X=arange(12).reshape(3,4), where rows are samples and columns are features.

```python
import torch
X=torch.arange(12,dtype=torch.float32).reshape(3,4)
print(X,X.shape,torch.cat([X,X],dim=0).shape)
```
