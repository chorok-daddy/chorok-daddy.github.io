## Logits and probabilities {#logit확률}

### Starting conditions {#시작-조건}

The inputs are the provided 12×2 synthetic features; samples whose two features sum to more than 1 belong to Class 1.

### Procedure {#진행-방법}

1.  Run the code while distinguishing the values passed to BCEWithLogitsLoss from those after applying Sigmoid.
2.  For each sample, record the two features, ground truth, logit, probability, and decision at .5 in a table.
3.  Plot the loss curve and show the classes in the 2-dimensional data.

### Results to retain {#남길-결과}

- A 12-row output table and plots of the loss and data.

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch import nn
X=torch.tensor([[-2.,-1.],[-1.,-2.],[-1.,0.],[0.,-1.],[1.,1.],[1.,2.],[2.,1.],[2.,2.],[-2.,1.],[1.,-2.],[0.,2.],[2.,0.]])
y=(X.sum(dim=1)>1).float().reshape(-1,1)
def fit_binary(weights=None):
    torch.manual_seed(7);m=nn.Linear(2,1);opt=torch.optim.SGD(m.parameters(),lr=.1);history=[]
    for _ in range(200):
        losses=nn.functional.binary_cross_entropy_with_logits(m(X),y,reduction='none')
        loss=(losses if weights is None else losses*weights).mean()
        opt.zero_grad();loss.backward();opt.step();history.append(loss.item())
    return m,history
model,history=fit_binary();prob=torch.sigmoid(model(X)).detach().reshape(-1).numpy();truth=y.numpy().reshape(-1).astype(int)
print(prob)
```
