# Linear Regression
 

### Imports

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as func
import torch.optim as opt

from google.colab import drive
drive.mount('/content/drive')

# Fix the Seed
torch.manual_seed(1)
```

### 1 Data loading
| x | y |
|---|---|
| 1 | 1 |
| 2 | 2 |
| 3 | 3 |



```python
# Enter Data manually
x_train = torch.FloatTensor([[1],[2],[3]])
y_train = torch.FloatTensor([[1],[2],[3]])

print(x_train)
print(x_train.shape)
print(y_train)
print(y_train.shape)
```

<span id="hypothesis-가정-and-cost"></span>

### 2 Hypothesis and Cost

$$y_{hypo} = H(x) = w \cdot x + b$$

- Assuming linearity, the equations above must hold.
- For 3 data points, $$\begin{bmatrix} y_1 \\ y_2 \\ y_3 \end{bmatrix} =  w \cdot \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix} + b$$
  - Find $w$ and $b$ that satisfy this equation.
- Use mean squared error as the cost function: $$\frac{1}{N} \sum_{n=1}^{n=N} (y_{hypo, (n)} - y_{train, (n)})^2$$

```python
# Initialize the Model (input dim, output dim)
model = nn.Linear(1, 1)

# Define the Hypothesis
y_hypo = model(x_train)
print(list(model.parameters()))
print(y_hypo)

# Compute Cost(Mean Sqaure Error)
cost = func.mse_loss(y_hypo, y_train)
print(cost)
```

<span id="cost-optimization"></span>

### 3 Cost Optimization

- Use stochastic gradient descent as the optimizer.

```python
# Configure the Optimizer (set learning rate = 0.01)
optimizer = opt.SGD(model.parameters(), lr=0.01)
```

```python
# Repeat
for epoch in range(1000):

  # Compute Cost / mse_loss(predicted value, true value)
  y_hypo = model(x_train)
  cost = func.mse_loss(y_hypo, y_train)

  # Update the model using cost
  optimizer.zero_grad()
  cost.backward()
  optimizer.step()

  # Print intermediate results every 100 iterations
  if epoch % 100 == 99:
    params = list(model.parameters())
    w = params[0].item()
    b = params[1].item()
    print('Epoch {:4d}/{} w: {:.3f} b: {:.3f} Cost: {:.6f}'.format(
        epoch+1, 1000, w, b, cost.item()
    ))
```

<span id="multivariable-linear-regression"></span>

### 4 Multivariable Linear Regression

$$y_{hypo} = H(\textbf{x}) = \textbf{W} \cdot \textbf{x} + b$$

- Under the linearity assumption, the equation above must hold: $$y_{hypo} = \begin{bmatrix} w_{1} && w_{2} && w_{3} \end{bmatrix}  \cdot \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix} + \begin{bmatrix} b \end{bmatrix}$$
  - Find $w$ and $b$ that minimize the MSE cost.

```python
# Use Numpy to load a csv file as an ndarray
dataset = np.loadtxt(
    '/content/drive/MyDrive/Colab Notebooks/courses/ML_Exp/data_linear_regression.csv', 
    delimiter=',', 
    dtype=np.float32)
# Shuffle the order randomly
np.random.shuffle(dataset)

# Convert to a torch tensor
x_train = torch.FloatTensor(dataset[:,:-1])
y_train = torch.FloatTensor(dataset[:,[-1]])

# Linear model
model = nn.Linear(3,1)

# Configure the Optimizer
optimizer = opt.SGD(model.parameters(), lr=0.00001)

# Repeat
for epoch in range(1000):
  y_hypo = model(x_train)
  cost = func.mse_loss(y_hypo, y_train)

  # model update
  optimizer.zero_grad()
  cost.backward()
  optimizer.step()

  # Print intermediate results every 100 iterations
  if epoch % 100 == 99:
    print('Epoch {:4d}/{} Cost: {:.6f}'.format(epoch+1, 1000, cost.item()))

print(y_hypo-y_train)
```
