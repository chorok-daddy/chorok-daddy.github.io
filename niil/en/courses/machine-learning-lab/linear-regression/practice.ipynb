## Sensor calibration with bias

<span id="시작-조건"></span>

### Starting conditions

The provided code generates synthetic data by adding small noise to y=1.8x+.7 at $x\in\{0,.5\}$,…,3.5. Use `seed=7`.

<span id="진행-방법"></span>

### Procedure

1.  Run the provided 200-epoch baseline example and plot the loss curve and prediction line.
2.  Record the learned w and b separately from the data generation coefficients in a table.
3.  Given the noise, explain whether the learned coefficients should exactly match the data generation coefficients.

<span id="남길-결과"></span>

### Results to retain

- Plots of the data, prediction line, and loss, plus a coefficient table.

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch import nn
torch.manual_seed(7)
x=torch.arange(0,4,.5).reshape(-1,1)
y=1.8*x+.7+.15*torch.randn_like(x)
def fit_regression(lr=.05,bias=True,epochs=200):
    torch.manual_seed(7);model=nn.Linear(1,1,bias=bias);opt=torch.optim.SGD(model.parameters(),lr=lr);history=[]
    for _ in range(epochs):
        loss=((model(x)-y)**2).mean();opt.zero_grad();loss.backward();opt.step();history.append(loss.item())
    return model,history
model,history=fit_regression()
print('weight',model.weight.detach().numpy(),'bias',model.bias.detach().numpy())
```
