## 풀이 예제

<!-- exercise-guidance:intro-machine-learning/week-04/learning-example -->
점수=[45,60,75,90]이고 합격 기준은 60 이상입니다.
<!-- /exercise-guidance:intro-machine-learning/week-04/learning-example -->


```python
scores=[45,60,75,90]
def pass_flags(values,threshold=60):
    return [v>=threshold for v in values]
print(pass_flags(scores))
```
