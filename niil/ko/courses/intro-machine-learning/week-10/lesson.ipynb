Random Forest는 여러 트리의 결과를 모읍니다. 단일 트리와 같은 split·전처리·지표로 비교해야 복잡도 차이를 해석할 수 있습니다. 정확도와 클래스별 오류를 함께 봅니다.

## 풀이 예제

<!-- exercise-guidance:intro-machine-learning/week-10/learning-example -->
Iris 데이터를 test_size=.3, random_state=7, stratify=y로 나눕니다. 깊이는 1,3,None입니다.
<!-- /exercise-guidance:intro-machine-learning/week-10/learning-example -->


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
X,y=load_iris(return_X_y=True)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=7,stratify=y)
model=DecisionTreeClassifier(max_depth=3,random_state=7).fit(Xtr,ytr)
print(model.score(Xte,yte))
```
