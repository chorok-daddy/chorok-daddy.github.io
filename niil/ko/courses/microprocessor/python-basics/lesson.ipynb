## 풀이 예제

<!-- exercise-guidance:microprocessor/python-basics/learning-example -->
기록은 [(0,21.0),(1,21.5),(2,None),(3,22.0),(4,20.5)]이며, 각 쌍은 시간[s]·온도[°C]입니다.
<!-- /exercise-guidance:microprocessor/python-basics/learning-example -->


```python
records=[(0,21.),(1,21.5),(2,None),(3,22.),(4,20.5)]
valid=[value for _,value in records if value is not None]
print('valid count',len(valid),'mean',sum(valid)/len(valid))
```
