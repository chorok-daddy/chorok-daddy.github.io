# Notebook 기초 · 학습


## 이번에 배울 내용

<!-- exercise-guidance:robot-basic-lab/notebook-basics/learning-intro -->
코드만 남기면 입력과 판단 근거를 잊기 쉽습니다. 작은 실험을 다시 읽고 확인할 수 있는 기록으로 만듭니다.
<!-- /exercise-guidance:robot-basic-lab/notebook-basics/learning-intro -->


## 전체 흐름

<!-- exercise-guidance:robot-basic-lab/notebook-basics/learning-flow -->
입력과 단위 → 코드·출력 → 설명·그림 → 처음부터 재실행
<!-- /exercise-guidance:robot-basic-lab/notebook-basics/learning-flow -->


## 계산과 설명을 한 기록에 두기

<!-- exercise-guidance:robot-basic-lab/notebook-basics/learning-1 -->
코드는 계산을, Markdown은 입력의 의미·단위·판단 근거를 남깁니다. 결과 숫자만 붙이지 말고 무엇을 넣었는지 바로 위에 적습니다. 표와 그래프에는 조건·축·단위를 표시합니다.
<!-- /exercise-guidance:robot-basic-lab/notebook-basics/learning-1 -->


## 다시 실행할 수 있는 기록

<!-- exercise-guidance:robot-basic-lab/notebook-basics/learning-2 -->
Notebook은 셀을 여러 순서로 실행할 수 있어 화면에 없는 변수에 의존하기 쉽습니다. 완성 후에는 Kernel을 다시 시작하고 위에서부터 실행합니다. 바꾼 조건과 마지막 출력이 같은 실험을 가리키는지 확인합니다.
<!-- /exercise-guidance:robot-basic-lab/notebook-basics/learning-2 -->


## 풀이 예제

<!-- exercise-guidance:robot-basic-lab/notebook-basics/learning-example -->
시간 0·1·2초에 거리 1.2·1.0·0.9 m를 기록했다면, 구간별 거리 변화율은 −0.2·−0.1 m/s입니다. 코드 앞에는 입력과 단위를, 결과 옆에는 음수의 의미를 적습니다.
<!-- /exercise-guidance:robot-basic-lab/notebook-basics/learning-example -->
