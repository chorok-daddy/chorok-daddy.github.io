## Commit은 비교 가능한 지점

<!-- exercise-guidance:robot-basic-lab/git-collaboration/learning-1 -->
파일을 저장하는 것과 Git Commit은 다릅니다. 작은 의미 단위로 변경을 기록하고 diff로 무엇이 바뀌었는지 읽습니다. Branch는 실험을 따로 진행할 이력의 이름입니다.
<!-- /exercise-guidance:robot-basic-lab/git-collaboration/learning-1 -->


## 충돌은 의도를 결정하는 문제

<!-- exercise-guidance:robot-basic-lab/git-collaboration/learning-2 -->
같은 줄이 다르게 바뀌면 Git이 올바른 최종 내용을 정할 수 없습니다. 두 변경의 목적을 읽고 원하는 내용을 남긴 뒤 다시 실행해 확인합니다. 단순히 충돌 표시만 지우는 것으로 검증이 끝나지 않습니다.
<!-- /exercise-guidance:robot-basic-lab/git-collaboration/learning-2 -->


## 풀이 예제

<!-- exercise-guidance:robot-basic-lab/git-collaboration/learning-example -->
코드가 만든 임시 저장소의 README.md에서 Condition: baseline을 Condition: test로 바꿉니다.
<!-- /exercise-guidance:robot-basic-lab/git-collaboration/learning-example -->


```python
from pathlib import Path
import tempfile, subprocess
lab_tmp=tempfile.TemporaryDirectory(prefix='course-git-')
repo=Path(lab_tmp.name)
def git(*args):
    return subprocess.check_output(['git','-C',str(repo),*args],text=True).strip()
git('init','-b','main');git('config','user.name','Learner');git('config','user.email','learner@example.invalid')
(repo/'README.md').write_text('Robot experiment\nCondition: baseline\n')
git('add','README.md');git('commit','-m','Record baseline')
print(repo,git('log','--oneline'))
```
