## 풀이 예제

<!-- exercise-guidance:machine-learning-lab/python-review/learning-example -->
3특징 자료 rows=[[1,2,3],[2,None,4],[3,4,5],[4,5,6]]입니다. None이 있는 행은 별도 목록으로 둡니다.
<!-- /exercise-guidance:machine-learning-lab/python-review/learning-example -->


```python
import numpy as np
import matplotlib.pyplot as plt
rows=[[1,2,3],[2,None,4],[3,4,5],[4,5,6]]
complete=[row for row in rows if None not in row]
X=np.asarray(complete,dtype=float)
print(X.shape,X.mean(axis=0))
```
