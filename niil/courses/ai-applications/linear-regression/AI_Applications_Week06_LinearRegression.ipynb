## 1. 원점을 지나는 회귀 문제

A의 첫 열은 입력 x, 두 번째 열의 1은 Bias용 상수입니다. 이번 기본 문제는 Bias를 고정하므로 첫 열만 A_prime으로 사용합니다. Hypothesis는 model(A_prime), Cost는 MSE입니다.

### A. 기준 예제 · 먼저 읽고 실행


````python
import sys, numpy as np, matplotlib.pyplot as plt
import torch
from torch import nn
import torch.nn.functional as F
torch.manual_seed(42)
torch.set_num_threads(2)
print('Python:', sys.version.split()[0], '| PyTorch:', torch.__version__)
A=torch.tensor([[0.,1.],[1.,1.],[2.,1.],[3.,1.]])
B=torch.tensor([[-1.],[0.2],[0.9],[2.1]])
A_prime=A[:,0:1]; B_prime=B
print(A_prime.shape,B_prime.shape)
plt.scatter(A_prime.numpy(),B_prime.numpy());plt.xlabel('x');plt.ylabel('y');plt.show()
````

## 2. 학습과 기록

train_linear(lr=0.01, epochs=1000)은 새 모델을 같은 Seed로 만들어 학습합니다. Learning Rate 비교에서는 lr만 바꾸고 epochs는 고정하세요. Gradient는 Loss가 증가하는 방향이며 SGD는 반대 방향으로 Weight를 갱신합니다.

### A. 기준 예제 · 먼저 읽고 실행


````python
def train_linear(lr=0.01,epochs=1000):
    torch.manual_seed(42);model=nn.Linear(1,1,bias=False)
    opt=torch.optim.SGD(model.parameters(),lr=lr);history=[]
    for _ in range(epochs):
        pred=model(A_prime);loss=F.mse_loss(pred,B_prime)
        opt.zero_grad();loss.backward();opt.step();history.append(loss.item())
    return model,history
model,history=train_linear()
print('w:',model.weight.detach(),'MSE:',F.mse_loss(model(A_prime),B_prime).item())
plt.plot(history);plt.xlabel('Epoch');plt.ylabel('MSE');plt.grid();plt.show()
````

## 3. 직접 해와 대조

pinv는 일반적인 Pseudoinverse(의사역행렬) 계산입니다.

### A. 기준 예제 · 먼저 읽고 실행

````python
w_pseudo=torch.linalg.pinv(A_prime)@B_prime
w_gradient=model.weight.detach().T
print('pinv:',w_pseudo,'gradient:',w_gradient,'difference:',(w_pseudo-w_gradient).abs())
plt.scatter(A_prime.numpy(),B_prime.numpy(),label='data')
plt.plot(A_prime.numpy(),model(A_prime).detach().numpy(),label='fitted');plt.legend();plt.show()
assert torch.allclose(w_pseudo,w_gradient,atol=1e-3)
````
