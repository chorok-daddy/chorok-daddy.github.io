## 식을 정리해도 값이 같은지 확인하기

<!-- exercise-guidance:control-2/laplace-transform/activity-1 -->
### 시작 조건

D=(x²−2x+1)/(x²−3x+2)를 약분합니다. 값 비교에는 x=0,3을 사용합니다.

### 진행 방법

1. D를 약분하고, 원래 분모가 0이 되는 x를 따로 표시하세요.
2. x=0,3에서 약분 전후 값이 같은지 확인하세요. x=1,2는 왜 같은 방식으로 확인할 수 없는지 설명합니다.

### 남길 결과

- 약분 전후 식과 제외할 값, 두 점의 대입 결과.
<!-- /exercise-guidance:control-2/laplace-transform/activity-1 -->


```python
import sympy as sym
x=sym.symbols('x')
D=(x**2-2*x+1)/(x**2-3*x+2)
print('약분:',sym.cancel(D))
print('원래 분모의 근:',sym.solve(x**2-3*x+2,x))
# TODO: x=0,3의 값을 대조하고 제외한 두 점을 설명하세요.
```

## Laplace Transform과 역변환

<!-- exercise-guidance:control-2/laplace-transform/activity-2 -->
Laplace Transform과 역변환을 차례로 구하세요.

### 시작 조건

t>0에서 f(t)=exp(−2t)를 변환하고 다시 역변환합니다.

### 진행 방법

1. sym.laplace_transform과 inverse_laplace_transform으로 변환·역변환을 구하세요.
2. t=.5,1,2에서 원래 함수와 복원한 함수의 값을 대조하세요.
3. 변환식의 Pole −2와 시간 영역의 감쇠 항 exp(−2t)를 연결해 설명하세요.

### 남길 결과

- F(s)·복원한 함수·수치 차이의 짧은 표.
<!-- /exercise-guidance:control-2/laplace-transform/activity-2 -->


```python
import sympy as sym
t=sym.symbols('t',positive=True);s=sym.symbols('s')
f=sym.exp(-2*t)
F=sym.laplace_transform(f,t,s,noconds=True)
restored=sym.inverse_laplace_transform(F,s,t)
print(F,restored)
# TODO: 지정한 세 시각에서 차이를 확인하세요.
```

## Initial Condition이 있는 Differential Equation

### 시작 조건

y′+2y=1, y(0)=0과 y(0)=1을 비교합니다. 입력은 t≥0의 단위 계단입니다.

### 진행 방법

1. Laplace 영역에서 (s+2)Y(s)=1/s+y(0)를 직접 쓰세요.
2. 두 Initial Condition의 역변환을 구해 t=0~4 s에 함께 그리세요.
3. 해를 원래 Differential Equation과 Initial Condition에 대입해 잔차를 확인하세요.

### 남길 결과

- 두 응답 곡선과 잔차 표, Initial Condition의 역할 짧은 설명.
<!-- /exercise-guidance:control-2/laplace-transform/activity-3 -->


```python
import sympy as sym
t=sym.symbols('t',positive=True); s=sym.symbols('s')
for y0 in [0,1]:
    Y=(1/s+y0)/(s+2)
    y=sym.inverse_laplace_transform(Y,s,t)
    print(y0,y)
# TODO: 잔차 확인과 그래프를 추가하세요.
```
