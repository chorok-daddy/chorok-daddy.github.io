G=(s²+10s+24)/(s²−8s+15), K=.5,1. 음의 단위 Feedback입니다.

### 진행 방법

1. Open-loop의 우반평면 Pole 수 P를 세세요.
2. 각 K의 Nyquist plot을 그리고 Closed-loop Pole의 우반평면 개수 Z를 직접 계산하세요.
3. -1점의 감김 방향에 대한 본인의 부호 규약을 적고 P·감김 수·Z의 관계를 확인하세요.
