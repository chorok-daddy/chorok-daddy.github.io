## Gain과 Closed-loop Pole

<!-- exercise-guidance:control-2/root-locus/activity-1 -->
### 시작 조건

L=(s+7)/(s(s+5)(s+15)(s+20)). K=100,500,1000을 사용합니다.

### 진행 방법

1. Root Locus를 그리고 각 K에서 feedback(K*L,1)의 Pole을 별도 기호로 표시하세요.
2. 각 Pole의 실수부와 안정 여부를 표에 기록하세요.
3. 기준 조건 K=500에 비해 Pole 이동과 응답 변화의 예상을 적으세요.

### 남길 결과

- Root Locus 그림 1개와 K별 Pole 표.
<!-- /exercise-guidance:control-2/root-locus/activity-1 -->


```python
import numpy as np
import matplotlib.pyplot as plt
import control as ctl
s = ctl.tf('s')
L=(s+7)/(s*(s+5)*(s+15)*(s+20))
ctl.root_locus_plot(L)
for k in [100,500,1000]: print(k,ctl.poles(ctl.feedback(k*L,1)))
```

## Time Response로 예상 확인

<!-- exercise-guidance:control-2/root-locus/activity-2 -->
### 시작 조건

같은 모델·Gain에 t=0~8 s의 Unit Step을 적용합니다.

### 진행 방법

1. 안정한 경우에만 Overshoot·Settling Time을 구하세요.
2. 가장 느린 Pole 쌍과 Time Response의 진동·감쇠를 연결해 설명하세요.
3. K=500보다 크거나 작은 값 하나를 추가하고 예상을 검증하세요.

### 남길 결과

- 4조건까지 포함한 응답 그림과 성능 표.
<!-- /exercise-guidance:control-2/root-locus/activity-2 -->


```python
import numpy as np
import matplotlib.pyplot as plt
import control as ctl
s = ctl.tf('s')
L=(s+7)/(s*(s+5)*(s+15)*(s+20))
t=np.linspace(0,8,1601)
T=ctl.feedback(500*L,1)
t,y=ctl.step_response(T,t)
plt.plot(t,y);plt.xlabel('Time [s]');plt.grid()
# TODO: 나머지 Gain을 추가하세요.
```
