### 시작 조건

G=1/(s²+.5s+1), u=sin(ωt), ω=.2,1,3 rad/s. t=0~200 s, Δt=.01 s입니다.

### 진행 방법

1. forced_response로 출력을 계산하고 마지막 두 주기만 골라 입력과 출력을 그리세요.
2. 출력을 a*sin(ωt)+b*cos(ωt)에 최소제곱 적합해 이득 sqrt(a²+b²)와 위상 atan2(b,a)를 구하세요.
3. frequency_response가 주는 이득·위상과 같은 단위로 비교하세요.

### 남길 결과

- 주파수 3행 표와 마지막 두 주기의 비교 그림 3개.
<!-- /exercise-guidance:control-2/frequency-response/activity-1 -->


```python
import numpy as np
import matplotlib.pyplot as plt
import control as ctl
s = ctl.tf('s')
G=1/(s*s+.5*s+1)
omega=1.;t=np.arange(0,200,.01);u=np.sin(omega*t)
response=ctl.forced_response(G,t,u)
y=np.asarray(response.outputs).reshape(-1)
keep=t>=t[-1]-4*np.pi/omega
# TODO: keep 구간의 a,b를 적합하고 이득·위상을 계산하세요.
```

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ctl
s = ctl.tf('s')
omega=np.logspace(-2,2,400)
for a in [.2,.5,1.]:
    response=ctl.frequency_response(1/(s*s+a*s+1),omega)
    print(a,np.max(response.magnitude))
# TODO: 위상과 공진 주파수 표를 완성하세요.
```
