## 직렬·병렬·Feedback 대조

<!-- exercise-guidance:control-2/transfer-functions/activity-2 -->
### 시작 조건

$H_1(s)=\frac2s$, $H_2(s)=\frac3{4s+1}$. Feedback은 $H_1H_2$에 대한 음의 단위 Feedback입니다.

### 진행 방법

1. $H_s=H_1H_2$, $H_p=H_1+H_2$, $T=\frac{H_s}{1+H_s}$를 식으로 구하세요.
2. ctl.series, ctl.parallel, ctl.feedback로 같은 모델을 만들고, ω=[0.1,1,10] rad/s에서 직접 만든 식과 비교하세요.
3. T의 Unit Step을 $0\le t\le 50$ s에서 그리세요. Hs·Hp의 원점 Pole과 DC Gain의 의미를 구분합니다.

### 남길 결과

- 연결 3종의 계수 표 1개와 T의 Step plot 1개.
- 오차는 같은 주파수에서 복소 응답 차이의 절댓값으로 기록합니다.
<!-- /exercise-guidance:control-2/transfer-functions/activity-2 -->


```python
import numpy as np
import matplotlib.pyplot as plt
import control as ctl
s = ctl.tf('s')
H1=ctl.tf([2],[1,0]); H2=ctl.tf([3],[4,1])
Hs=ctl.series(H1,H2); Hp=ctl.parallel(H1,H2); T=ctl.feedback(Hs,1)
print(Hs, Hp, T)
t=np.linspace(0,50,2501)
t,y=ctl.step_response(T,t)
plt.plot(t,y); plt.xlabel('Time [s]'); plt.ylabel('Output'); plt.grid()
# TODO: 직접 계산한 모델과 주파수응답을 비교하세요.
```
