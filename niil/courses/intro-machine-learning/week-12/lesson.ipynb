## 풀이 예제

<!-- exercise-guidance:intro-machine-learning/week-12/learning-example -->
Iris를 위와 같은 70/30 분할로 나누고 Train에서만 StandardScaler를 fit합니다.
<!-- /exercise-guidance:intro-machine-learning/week-12/learning-example -->


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
X,y=load_iris(return_X_y=True)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=7,stratify=y)
scaler=StandardScaler().fit(Xtr);Xtr=scaler.transform(Xtr);Xte=scaler.transform(Xte)
```

<!-- exercise-guidance:intro-machine-learning/week-12/learning-worked-model -->
위에서 준비한 분할·Standardization 결과로 한 모델을 학습하고 평가합니다.
<!-- /exercise-guidance:intro-machine-learning/week-12/learning-worked-model -->


```python
model = LogisticRegression(max_iter=500)
model.fit(Xtr, ytr)
prediction = model.predict(Xte)
print("Train Accuracy:", model.score(Xtr, ytr))
print("Test Accuracy:", model.score(Xte, yte))
```
