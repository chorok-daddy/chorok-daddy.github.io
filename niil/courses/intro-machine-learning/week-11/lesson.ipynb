## 정답 없이 묶기

<!-- exercise-guidance:intro-machine-learning/week-11/learning-1 -->
K-means는 정한 군집 수에서 중심과 표본의 거리를 줄입니다. 군집 번호는 임의의 이름이라 번호 자체보다 중심·분산·배정 구조를 봅니다.

## 거리의 단위

<!-- exercise-guidance:intro-machine-learning/week-11/learning-2 -->
한 특징의 값 범위가 크면 그 특징이 거리를 지배할 수 있습니다. Standardization 전후와 군집 수를 비교하되, 관성값 감소만으로 의미 있는 군집을 얻었다고 판단하지 않습니다.
<!-- /exercise-guidance:intro-machine-learning/week-11/learning-2 -->


## 풀이 예제

<!-- exercise-guidance:intro-machine-learning/week-11/learning-example -->
make_blobs(n_samples=180,centers=3,cluster_std=.8,random_state=7), $K\in\{2,3,4\}$를 사용합니다.
<!-- /exercise-guidance:intro-machine-learning/week-11/learning-example -->


```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
X,_=make_blobs(n_samples=180,centers=3,cluster_std=.8,random_state=7)
model=KMeans(n_clusters=3,random_state=7,n_init=10).fit(X)
print(model.inertia_,silhouette_score(X,model.labels_))
```
