### 시작 조건

외부 다운로드 없는 sklearn load_digits의 8×8 영상 중 Class 0~4를 사용합니다. 분할은 stratified 70/30, `seed=7`입니다.

### 진행 방법

1. 영상 한 장의 8×8 배열과 Flatten된 64개 특징을 확인하세요.
2. Train에 fit한 StandardScaler와 LogisticRegression(max_iter=1000)으로 기준 분류를 수행하세요.

### 남길 결과

- 데이터 구조 표, 예시 영상 패널과 Test Accuracy.
<!-- /exercise-guidance:intro-machine-learning/week-13/activity-1 -->


```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
D=load_digits();mask=D.target<5;X=D.data[mask];y=D.target[mask]
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=7,stratify=y)
scaler=StandardScaler().fit(Xtr);clf=LogisticRegression(max_iter=1000).fit(scaler.transform(Xtr),ytr)
print(clf.score(scaler.transform(Xte),yte))
```
