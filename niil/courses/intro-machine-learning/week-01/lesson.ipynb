## 풀이 예제

<!-- exercise-guidance:intro-machine-learning/week-01/learning-example -->
x=[1,2,3,4,5,6,7,8], y=[0,0,0,0,1,1,1,1]입니다. x는 센서 특징, y는 두 상태의 Label입니다.
<!-- /exercise-guidance:intro-machine-learning/week-01/learning-example -->


```python
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(1,9);y=(x>4).astype(int)
pred=(x>4).astype(int)
print(np.mean(pred==y))
```
