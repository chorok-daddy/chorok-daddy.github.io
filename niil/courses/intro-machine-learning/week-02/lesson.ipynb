## 풀이 예제

<!-- exercise-guidance:intro-machine-learning/week-02/learning-example -->
8×8 영상 A는 중앙 두 열, B는 중앙 두 행만 1이고 나머지는 0입니다.
<!-- /exercise-guidance:intro-machine-learning/week-02/learning-example -->


```python
import numpy as np
import matplotlib.pyplot as plt
A=np.zeros((8,8));B=A.copy();A[:,3:5]=1;B[3:5,:]=1
print(A.mean(),B.mean(),A.sum(axis=0),B.sum(axis=0))
import matplotlib.pyplot as plt
fig,axes=plt.subplots(1,2,figsize=(6,3))
axes[0].imshow(A,cmap="gray",vmin=0,vmax=1);axes[0].set_title("A")
axes[1].imshow(B,cmap="gray",vmin=0,vmax=1);axes[1].set_title("B")
```
