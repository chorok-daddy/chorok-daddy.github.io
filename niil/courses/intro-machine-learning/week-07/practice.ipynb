### 시작 조건

x=[0,1,2,3,4,5], y=[1,3.2,4.8,7.1,8.9,11.2]입니다. 모델은 y_hat=w*x+b입니다.

### 진행 방법

1. w=2,b=1인 예측·잔차·MSE를 구하세요.
2. np.linalg.lstsq로 구한 w,b와 기준 모델을 같은 표에서 비교하세요.

### 남길 결과

- 예측·산점도 그림과 모델 비교 표.
<!-- /exercise-guidance:intro-machine-learning/week-07/activity-1 -->


```python
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(6.);y=np.array([1,3.2,4.8,7.1,8.9,11.2]);X=np.c_[x,np.ones_like(x)]
coef=np.linalg.lstsq(X,y,rcond=None)[0]
print(coef)
```
