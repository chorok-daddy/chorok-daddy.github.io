### 진행 방법

2. np.linalg.lstsq로 구한 w,b와 기준 모델을 같은 표에서 비교하세요.

### 남길 결과

- 예측·산점도 그림과 모델 비교 표.

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(6.);y=np.array([1,3.2,4.8,7.1,8.9,11.2]);X=np.c_[x,np.ones_like(x)]
coef=np.linalg.lstsq(X,y,rcond=None)[0]
print(coef)
```
