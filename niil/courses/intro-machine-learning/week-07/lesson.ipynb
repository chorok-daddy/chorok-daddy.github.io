## 풀이 예제

<!-- exercise-guidance:intro-machine-learning/week-07/learning-example -->
$x=\begin{bmatrix}0&1&2&3&4&5\end{bmatrix}$, $y=\begin{bmatrix}1&3.2&4.8&7.1&8.9&11.2\end{bmatrix}$입니다. 모델은 y_hat=w*x+b입니다.
<!-- /exercise-guidance:intro-machine-learning/week-07/learning-example -->


```python
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(6.);y=np.array([1,3.2,4.8,7.1,8.9,11.2]);X=np.c_[x,np.ones_like(x)]
coef=np.linalg.lstsq(X,y,rcond=None)[0]
print(coef)
```
