## Bias가 있는 센서 보정

<!-- exercise-guidance:machine-learning-lab/linear-regression/activity-1 -->
### 시작 조건

제공 코드는 x=0,.5,...,3.5에서 y=1.8x+.7에 작은 노이즈를 더한 합성 자료를 만듭니다. seed=7입니다.

### 진행 방법

1. 제공한 200 epoch 기준 예제를 실행하고 Loss 곡선과 예측 직선을 그리세요.
2. 학습된 w·b와 데이터 생성 계수를 나누어 표에 적으세요.
3. 노이즈가 있으므로 학습 계수가 생성 계수와 완전히 같아야 하는지 설명하세요.

### 남길 결과

- 데이터·예측 직선과 Loss 그림, 계수 표.
<!-- /exercise-guidance:machine-learning-lab/linear-regression/activity-1 -->


```python
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch import nn
torch.manual_seed(7)
x=torch.arange(0,4,.5).reshape(-1,1)
y=1.8*x+.7+.15*torch.randn_like(x)
def fit_regression(lr=.05,bias=True,epochs=200):
    torch.manual_seed(7);model=nn.Linear(1,1,bias=bias);opt=torch.optim.SGD(model.parameters(),lr=lr);history=[]
    for _ in range(epochs):
        loss=((model(x)-y)**2).mean();opt.zero_grad();loss.backward();opt.step();history.append(loss.item())
    return model,history
model,history=fit_regression()
print('weight',model.weight.detach().numpy(),'bias',model.bias.detach().numpy())
```
