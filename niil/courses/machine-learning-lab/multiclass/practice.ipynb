## Multiclass Logits와 Prediction 비교

<!-- exercise-guidance:machine-learning-lab/multiclass/activity-1 -->
### 시작 조건

제공된 중심 (-2,0),(2,0),(0,2) 주변에서 Class별 6표본, 총 18개를 만듭니다. seed=7입니다.

### 진행 방법

1. Linear(2,3)와 CrossEntropy의 입력·Label shape를 확인하세요.
2. 첫 6개 표본의 Logit 3개·확률 3개·예측을 기록하세요.
3. 자료 산점도와 Loss 곡선을 그리세요.

### 남길 결과

- 출력 6행 표와 자료·Loss 그림.
<!-- /exercise-guidance:machine-learning-lab/multiclass/activity-1 -->


```python
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch import nn
rng=np.random.default_rng(7)
centers=np.array([[-2.,0.],[2.,0.],[0.,2.]])
X_np=np.concatenate([c+rng.normal(0,.25,(6,2)) for c in centers]);y_np=np.repeat(np.arange(3),6)
X=torch.tensor(X_np,dtype=torch.float32);y=torch.tensor(y_np,dtype=torch.long)
def fit_multi(labels=y):
    torch.manual_seed(7);m=nn.Linear(2,3);opt=torch.optim.SGD(m.parameters(),lr=.1);history=[]
    for _ in range(200):
        loss=nn.functional.cross_entropy(m(X),labels);opt.zero_grad();loss.backward();opt.step();history.append(loss.item())
    return m,history
model,history=fit_multi();logits=model(X).detach();prob=torch.softmax(logits,dim=1);pred=prob.argmax(dim=1)
print(prob[:3],pred[:3])
```
