## 한 파일의 변경 추적

<!-- exercise-guidance:microprocessor/git-collaboration/activity-1 -->
### 시작 조건

`repo`는 Notebook 커널이 실행되는 컴퓨터의 임시 저장소입니다. 그 안의 `README.md`에서 `Condition: baseline`을 `Condition: test`로 바꿉니다.

### 진행 방법

1. 기준 Commit을 확인한 뒤 같은 Notebook의 새 코드 셀에서 `repo/'README.md'`를 읽고 해당 문장 한 줄만 바꾸어 저장하세요. 파일 경로는 `repo`를 기준으로 사용합니다.
2. 제공된 `git(*args)` 함수를 코드 셀에서 호출해 diff와 status를 기록하세요. 예를 들어 `print(git('diff'))`로 변경 내용을 표시할 수 있습니다.
3. 변경을 Commit한 뒤 직전 Commit과의 차이를 다시 확인하세요.

### 남길 결과

- 수정 전·후·Commit 후의 3행 표와 diff 코드 블록.
<!-- /exercise-guidance:microprocessor/git-collaboration/activity-1 -->


```python
from pathlib import Path
import tempfile, subprocess
lab_tmp=tempfile.TemporaryDirectory(prefix='course-git-')
repo=Path(lab_tmp.name)
def git(*args):
    return subprocess.check_output(['git','-C',str(repo),*args],text=True).strip()
git('init','-b','main');git('config','user.name','Learner');git('config','user.email','learner@example.invalid')
(repo/'README.md').write_text('Robot experiment\nCondition: baseline\n')
git('add','README.md');git('commit','-m','Record baseline')
print(repo,git('log','--oneline'))
```
