## 한 Frame의 비용

<!-- exercise-guidance:microprocessor/camera/learning-1 -->
압축 전 저장량은 높이×너비×채널 수×채널당 Byte입니다. 해상도를 두 축에서 두 배로 키우면 픽셀 수는 네 배가 됩니다. ROI로 필요한 구역만 쓰면 관측 정보와 처리 비용을 함께 조절할 수 있습니다.
<!-- /exercise-guidance:microprocessor/camera/learning-1 -->


## 요청한 FPS와 실제 기록

<!-- exercise-guidance:microprocessor/camera/learning-2 -->

## 풀이 예제

<!-- exercise-guidance:microprocessor/camera/learning-example -->
제공 NumPy 코드가 만드는 120×160×3 uint8 RGB 영상을 사용합니다.
<!-- /exercise-guidance:microprocessor/camera/learning-example -->


```python
import numpy as np
import matplotlib.pyplot as plt
x,y=np.meshgrid(np.linspace(0,255,160),np.linspace(0,255,120))
frame=np.stack([x,y,255-x],axis=-1).astype(np.uint8)
small=frame[::2,::2]
print(frame.shape,frame.nbytes,small.shape,small.nbytes)
```
