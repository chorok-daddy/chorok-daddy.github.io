## 버튼의 물리 신호와 사건

<!-- exercise-guidance:microprocessor/gpio/learning-1 -->
기계식 접점은 한 번 눌러도 짧은 시간 동안 여러 번 바뀔 수 있습니다. 기록된 전환 시각을 보면 눌림 횟수와 신호 변화 횟수가 왜 다른지 알 수 있습니다.
<!-- /exercise-guidance:microprocessor/gpio/learning-1 -->


## 관측 간격과 Debounce

<!-- exercise-guidance:microprocessor/gpio/learning-2 -->
Polling은 값을 읽는 주기이고 Debounce는 가까운 변화를 묶는 규칙입니다. 너무 긴 Polling은 짧은 입력을 놓치고, 너무 긴 Debounce는 실제 연속 입력을 합칠 수 있습니다.

시간 단위를 통일해 비교합니다.
<!-- /exercise-guidance:microprocessor/gpio/learning-2 -->


## 풀이 예제

<!-- exercise-guidance:microprocessor/gpio/learning-example -->
(t_ms,level)=[(0,0),(10,1),(12,0),(14,1),(50,0),(80,1)]입니다. 두 기록 사이에는 직전 level이 유지됩니다.
<!-- /exercise-guidance:microprocessor/gpio/learning-example -->


```python
import numpy as np
import matplotlib.pyplot as plt
events=[(0,0),(10,1),(12,0),(14,1),(50,0),(80,1)]
rising=[t for (t,x),(_,previous) in zip(events[1:],events[:-1]) if previous==0 and x==1]
print('raw rising',rising)
```
