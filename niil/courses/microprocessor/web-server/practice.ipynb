## 요청을 함수로 시험하기

<!-- exercise-guidance:microprocessor/web-server/activity-1 -->
### 시작 조건

FastAPI의 /readings 경로에서 device와 start·end를 받습니다. TestClient는 실제 서버를 켜지 않고 요청을 보냅니다.

### 진행 방법

1. 제공된 정상 요청을 실행하고 Status code와 JSON 필드를 확인하세요.
2. start>end 요청을 실행하고 정상 응답과 구분하세요.

### 남길 결과

- 정상·오류 요청의 응답 예.
<!-- /exercise-guidance:microprocessor/web-server/activity-1 -->


```python
from fastapi import FastAPI,HTTPException
from fastapi.testclient import TestClient
app=FastAPI()
readings=[{'device':'m1','time_s':0.,'value':20.},{'device':'m1','time_s':1.,'value':21.}]
@app.get('/readings')
def get_readings(device:str,start:float=0,end:float=10):
    if start>end:raise HTTPException(status_code=400,detail='start must not exceed end')
    return [row for row in readings if row['device']==device and start<=row['time_s']<=end]
client=TestClient(app)
for query in ['/readings?device=m1','/readings?device=m1&start=2&end=1']:
    result=client.get(query);print(result.status_code,result.json())
```
