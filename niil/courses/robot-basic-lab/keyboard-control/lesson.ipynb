## 입력 사건에서 상태로

<!-- exercise-guidance:robot-basic-lab/keyboard-control/learning-1 -->
키를 누른 사건은 속도 명령을 설정하고, 놓은 사건은 해제합니다. 이벤트를 받을 때마다 거리를 더하면 반복 입력 횟수와 처리 속도에 따라 움직임이 달라질 수 있습니다. 상태를 유지하고 시간 간격으로 위치를 적분합니다.
<!-- /exercise-guidance:robot-basic-lab/keyboard-control/learning-1 -->


## 명령의 유효 시간

<!-- exercise-guidance:robot-basic-lab/keyboard-control/learning-2 -->
