## Markdown 실습 기록

<!-- exercise-guidance:robot-basic-lab/notebook-basics/activity-1 -->
### 시작 조건

주제는 가상의 거리 센서입니다. $t=\begin{bmatrix}0&1&2\end{bmatrix}$ s, 거리=[1.2,1.0,.9] m를 사용합니다.

### 진행 방법

1. 제목 3단계와 실습 목적 2~3문장을 작성하세요.
2. 시간·거리·단위를 포함한 표와 Δ거리/Δ시간 수식 한 개를 넣으세요.
3. 관측 사실 두 개와 아직 판단할 수 없는 것 한 개를 구분하세요.

### 남길 결과

- Markdown 셀 1개: 제목·표·수식·해석을 포함한 한 화면 분량.

## 실행 기록과 예상 구분

<!-- exercise-guidance:robot-basic-lab/notebook-basics/activity-2 -->
### 시작 조건

같은 자료에서 0~1 s와 1~2 s의 평균 변화율을 구합니다.

### 진행 방법

1. 계산 전 두 구간의 감소 속도를 예상해 문장으로 적으세요.
2. 수식 대입과 계산 값을 기록하고 예상과 비교하세요.
3. 인용한 원본 학습자료 링크 하나와 참고한 부분을 짧게 적으세요.

### 남길 결과

- 두 구간의 표와 계산 과정 2줄.
