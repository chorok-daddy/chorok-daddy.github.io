## 배열에서 영상으로

<!-- exercise-guidance:robot-basic-lab/image-processing/activity-1 -->
### 시작 조건

120×160 BGR 영상에 파란 사각형을 그리고 RGB·Gray·HSV로 변환합니다.

### 진행 방법

1. 원본 배열의 shape·dtype·범위를 확인하세요.
2. 같은 영상을 Matplotlib에 올바르게 표시하고 채널 순서를 일부러 바꾼 결과와 비교하세요.

### 남길 결과

- 표현별 비교 그림과 배열 요약.
<!-- /exercise-guidance:robot-basic-lab/image-processing/activity-1 -->


```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
image=np.full((120,160,3),220,dtype=np.uint8);image[30:90,50:110]=(255,30,30)
rgb=cv.cvtColor(image,cv.COLOR_BGR2RGB);hsv=cv.cvtColor(image,cv.COLOR_BGR2HSV)
fig,axes=plt.subplots(1,2);axes[0].imshow(image);axes[0].set_title('BGR as RGB');axes[1].imshow(rgb);axes[1].set_title('Correct RGB')
print(image.shape,image.dtype,hsv[60,80])
```
