## 세 기법의 다른 위치

<!-- exercise-guidance:reinforcement-learning/week-04/learning-1 -->
Double은 다음 행동의 선택과 가치 평가, Dueling은 Q의 표현, PER는 경험의 샘플링을 바꿉니다. 모두 같은 문제를 해결하는 기법은 아닙니다.
<!-- /exercise-guidance:reinforcement-learning/week-04/learning-1 -->


## 표현과 분포

<!-- exercise-guidance:reinforcement-learning/week-04/learning-2 -->
Dueling은 $Q=V+A-\operatorname{mean}(A)$로 결합할 수 있습니다. PER는 큰 오차의 경험을 자주 고르는 대신 샘플 편향과 가중치를 함께 고려합니다. 경험의 범위 밖 상태를 만들어 주지는 않습니다.
<!-- /exercise-guidance:reinforcement-learning/week-04/learning-2 -->


## 풀이 예제

<!-- exercise-guidance:reinforcement-learning/week-04/learning-example -->
- $Q_{\mathrm{online}}=[2,4,3]$, $Q_{\mathrm{target}}=[5,1,2]$.
- $r=1$, $\gamma=.95$.
- Dueling: $V=2$, $A=[1,2,6]$.
- TD Error: $[0.1,1,5]$.
<!-- /exercise-guidance:reinforcement-learning/week-04/learning-example -->


```python
import numpy as np
online=np.array([2.,4.,3.]);target=np.array([5.,1.,2.])
print('DQN',1+.95*target.max(),'Double',1+.95*target[online.argmax()])
advantage=np.array([1.,2.,6.]);print('Dueling Q',2+advantage-advantage.mean())
error=np.array([.1,1.,5.])
for alpha in [0,.6]:
    probability=(error+1e-5)**alpha;probability/=probability.sum();print(alpha,probability)
```
